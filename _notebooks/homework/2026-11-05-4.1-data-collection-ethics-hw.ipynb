**Case Study Chosen:** Amazon's experimental résumé-screening tool, reported by Reuters in 2018.

**1. System Background.** Amazon built a machine learning tool to rank job applicants' résumés, so recruiters could focus on the strongest candidates without reading every application.

**2. Root Cause Analysis.** The model learned from about ten years of résumés submitted to the company, and most came from men, reflecting the tech industry's gender imbalance. It treated those past patterns as signs of a good candidate, so it reportedly penalized résumés containing the word "women's," as in "women's chess club captain." The data set was unfit for the goal: it recorded who had been hired before, not who would do the job well. Amazon ended the project.

**3. Developer Response.** As lead developer, I would first define success using job performance rather than past hiring decisions. I would audit the training data's representation before training, remove features that act as stand-ins for gender, and test ranking results across groups before and after any release. Because a model can find new stand-ins, I would use it only to support recruiters rather than to reject anyone automatically.
