## Chapter 1: No Shortcuts Allowed

`==` is right for an `int`. A `String` is an object, so it needs `.equals()` to compare the text instead of the memory location.

```java
public class ExactSearch {
    public static int indexOfID(int[] ids, int targetID) {
        for (int i = 0; i < ids.length; i++) {
            if (ids[i] == targetID) return i;
        }
        return -1;
    }

    public static int indexOfName(String[] roster, String targetName) {
        for (int i = 0; i < roster.length; i++) {
            if (roster[i].equals(targetName)) return i;
        }
        return -1;
    }

    public static void main(String[] args) {
        System.out.println(indexOfID(Roster.studentIDs, 40881));         // 5
        System.out.println(indexOfID(Roster.studentIDs, 99999));         // -1
        System.out.println(indexOfName(Roster.names, "Marcus Webb"));    // 10
    }
}

ExactSearch.main(null);
```

## Chapter 2: Search Doesn't Mean "Equals"

The skeleton is the same as Chapter 1, but the `if` tests a condition rather than an exact value.

```java
public class ConditionSearch {
    public static int indexOfFirstBelow(int[] scores, int threshold) {
        for (int i = 0; i < scores.length; i++) {
            if (scores[i] < threshold) return i;
        }
        return -1;
    }

    public static void main(String[] args) {
        int idx = indexOfFirstBelow(Roster.quizScores, 60);
        System.out.println(Roster.names[idx] + " scored " + Roster.quizScores[idx]);
    }
}

ConditionSearch.main(null);
```

## Chapter 3: The Sign-In Sheet

The bound is `numSignedIn`, not `sheet.length`. The empty slots hold Java's default 0, so searching the whole array would wrongly report that ID 0 had signed in.

```java
public class SignInSheet {
    public static boolean isSignedIn(int[] sheet, int numSignedIn, int targetID) {
        for (int i = 0; i < numSignedIn; i++) {
            if (sheet[i] == targetID) return true;
        }
        return false;
    }

    public static void main(String[] args) {
        int[] signInSheet = new int[15];
        int numSignedIn = 8;
        int[] realIDs = {40217, 40593, 40108, 40772, 40364, 40881, 40255, 40940};
        for (int i = 0; i < realIDs.length; i++) {
            signInSheet[i] = realIDs[i];
        }

        System.out.println(isSignedIn(signInSheet, numSignedIn, 40940));  // true
        System.out.println(isSignedIn(signInSheet, numSignedIn, 0));      // false
    }
}

SignInSheet.main(null);
```

## Chapter 4: When You Don't Stop at the First One

There's no early `return` here, because every match is needed. Prediction: Diego (54), Priya (46), and Marcus (58).

```java
import java.util.ArrayList;

public class AllMatches {
    public static ArrayList<Integer> indexesBelow(int[] scores, int threshold) {
        ArrayList<Integer> matches = new ArrayList<>();
        for (int i = 0; i < scores.length; i++) {
            if (scores[i] < threshold) {
                matches.add(i);
            }
        }
        return matches;
    }

    public static void main(String[] args) {
        for (int i : indexesBelow(Roster.quizScores, 60)) {
            System.out.println(Roster.names[i] + ": " + Roster.quizScores[i]);
        }
    }
}

AllMatches.main(null);
```

## Capstone: The Duplicate ID Detector

`j` starts at `i + 1`. Starting at 0 would compare each ID to itself and report every real pair twice.

```java
public class DuplicateDetector {
    public static void findDuplicateIDs(int[] ids, String[] roster) {
        for (int i = 0; i < ids.length; i++) {
            for (int j = i + 1; j < ids.length; j++) {
                if (ids[i] == ids[j]) {
                    System.out.println("Duplicate " + ids[i] + " at " + i + " & " + j
                            + " (" + roster[i] + " & " + roster[j] + ")");
                }
            }
        }
    }

    public static void main(String[] args) {
        findDuplicateIDs(Roster.studentIDs, Roster.names);
    }
}

DuplicateDetector.main(null);
```
