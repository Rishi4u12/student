## Popcorn Hack 1: Stack vs Heap

Predictions, before running: changing `b` won't affect `a`, so `a` stays 10. Changing `array2` will affect `array1`, so both print 99.

```java
public class MemoryDemo {
    public static void main(String[] args) {
        int a = 10;
        int b = a;
        b = 20;
        System.out.println("Primitives (Stack):");
        System.out.println("a = " + a);
        System.out.println("b = " + b);

        int[] array1 = {1, 2, 3};
        int[] array2 = array1;
        array2[0] = 99;
        System.out.println("\nArrays (Heap):");
        System.out.println("array1[0] = " + array1[0]);
        System.out.println("array2[0] = " + array2[0]);
    }
}

MemoryDemo.main(null);
```

**Task 1.** `int b = a` copies the value 10 on the stack, so `a` and `b` are independent. `int[] array2 = array1` copies only a reference, so both variables point to the same array on the heap, and a change through one is visible through the other.

**Task 2.** The stack holds `a`, `b`, and the references `array1` and `array2`. The heap holds the single array object `{99, 2, 3}`.

## Popcorn Hack 2: Modifying vs Reassigning

Prediction, before running: `haveBirthday` changes John to 21 because it modifies the object itself. `reassignPerson` prints New Person inside the method, but John is still John, 21, afterward.

```java
public class PersonDemo {
    static class Person {
        String name;
        int age;

        Person(String name, int age) {
            this.name = name;
            this.age = age;
        }
    }

    public static void haveBirthday(Person p) {
        p.age = p.age + 1;
        System.out.println("Inside method: " + p.name + " is now " + p.age);
    }

    public static void reassignPerson(Person p) {
        p = new Person("New Person", 99);
        System.out.println("Inside reassign: " + p.name + " is " + p.age);
    }

    public static void main(String[] args) {
        Person john = new Person("John", 20);
        System.out.println("Before birthday: " + john.name + " is " + john.age);
        haveBirthday(john);
        System.out.println("After birthday: " + john.name + " is " + john.age);

        System.out.println("\nBefore reassign: " + john.name + " is " + john.age);
        reassignPerson(john);
        System.out.println("After reassign: " + john.name + " is " + john.age);
    }
}

PersonDemo.main(null);
```

**Question 1.** John's age is 21. The parameter `p` holds a copy of the reference to John's object on the heap, so `p.age` changes that same object.

**Question 2.** John is still named John and is 21. `p = new Person(...)` only points the method's local copy of the reference at a new heap object, and `john` on `main`'s stack doesn't change.

**Question 3.** Modifying contents goes through the reference and changes the shared heap object, so every reference sees it. Reassigning changes only which object that one stack variable points to.
