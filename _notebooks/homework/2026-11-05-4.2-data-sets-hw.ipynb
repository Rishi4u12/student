`length - 1` finds the last index without hardcoding it, so it still works if the array size changes.

```java
public class MyWeek {
    public static void main(String[] args) {
        int[] sleepHours = {8, 7, 6, 9, 7};
        String[] days = {"Mon", "Tue", "Wed", "Thu", "Fri"};

        System.out.println("Nights tracked: " + sleepHours.length);
        System.out.println("First night (" + days[0] + "): " + sleepHours[0]);

        int last = sleepHours.length - 1;
        System.out.println("Last night (" + days[last] + "): " + sleepHours[last]);

        sleepHours[2] = 8;
        System.out.println("After update, " + days[2] + ": " + sleepHours[2]);
    }
}

MyWeek.main(null);
```

## Homework Hack: Score Report

The loop condition is `i < quizzes.length`, because the last valid index is one less than the length. Reading index `quizzes.length` throws the exception that the `catch` block prints.

```java
public class ScoreReport {
    public static void main(String[] args) {
        final String STUDENT_NAME = "Rishabh Jha";
        int[] quizzes = {88, 94, 76, 91, 83};

        System.out.println(STUDENT_NAME + "'s quiz report");
        System.out.println("Quizzes: " + quizzes.length);
        System.out.println("Last score: " + quizzes[quizzes.length - 1]);

        for (int i = 0; i < quizzes.length; i++) {
            System.out.println("Index " + i + ": " + quizzes[i]);
        }

        try {
            System.out.println(quizzes[quizzes.length]);
        } catch (ArrayIndexOutOfBoundsException error) {
            System.out.println("Caught: " + error.getMessage());
        }
    }
}

ScoreReport.main(null);
```
