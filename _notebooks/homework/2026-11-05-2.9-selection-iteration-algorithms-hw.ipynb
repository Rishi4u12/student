## Popcorn Hack: Prime Check

A prime has exactly 2 divisors, 1 and itself, so the method counts them. Prediction: 2, 13, and 97 print `true`, while 1, 15, and 100 print `false`.

```java
public class PrimeCheck {
    public static boolean isPrime(int n) {
        if (n < 2) {
            return false;
        }
        int divisors = 0;
        for (int d = 1; d <= n; d++) {
            if (n % d == 0) {
                divisors++;
            }
        }
        return divisors == 2;
    }

    public static void main(String[] args) {
        System.out.println("2: " + isPrime(2));
        System.out.println("13: " + isPrime(13));
        System.out.println("97: " + isPrime(97));
        System.out.println("1: " + isPrime(1));
        System.out.println("15: " + isPrime(15));
        System.out.println("100: " + isPrime(100));
    }
}

PrimeCheck.main(null);
```

## Homework Hack: Even Digits

`% 10` reads the last digit and `/ 10` removes it. Prediction: 2847 has 3 even digits, 1357 has 0, and 2000 has 4, because 0 is even.

```java
public class EvenDigits {
    public static void countEvenDigits(int n) {
        int original = n;
        int evenDigits = 0;
        while (n > 0) {
            int digit = n % 10;
            if (digit % 2 == 0) {
                evenDigits++;
            }
            n = n / 10;
        }
        System.out.println(original + " even digits: " + evenDigits);
    }

    public static void main(String[] args) {
        countEvenDigits(2847);
        countEvenDigits(1357);
        countEvenDigits(2000);
    }
}

EvenDigits.main(null);
```
