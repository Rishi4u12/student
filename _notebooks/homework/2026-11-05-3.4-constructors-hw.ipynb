## Popcorn Hack: Pet Constructor

The parameters have the same names as the fields, so `this.` is needed to reach the fields.

```java
class Pet {
    String name;
    String type;

    Pet(String name, String type) {
        this.name = name;
        this.type = type;
    }
}

public class Main {
    public static void main(String[] args) {
        Pet p = new Pet("Buddy", "dog");
        System.out.println(p.name + " is a " + p.type);
    }
}

Main.main(null);
```

## Homework Hack: Book Constructor

Each `new Book(...)` call runs the constructor once, so the two objects get their own values.

```java
class Book {
    String title;
    String author;
    int pages;

    Book(String title, String author, int pages) {
        this.title = title;
        this.author = author;
        this.pages = pages;
    }
}

public class Main {
    public static void main(String[] args) {
        Book b1 = new Book("The Alchemist", "Paulo Coelho", 208);
        Book b2 = new Book("1984", "George Orwell", 328);
        System.out.println(b1.title + " by " + b1.author + ", " + b1.pages + " pages");
        System.out.println(b2.title + " by " + b2.author + ", " + b2.pages + " pages");
    }
}

Main.main(null);
```
