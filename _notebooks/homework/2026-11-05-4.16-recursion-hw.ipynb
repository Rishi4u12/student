## Popcorn Hack: Count Up

**Base case:** `number > 5`, which stops the recursion. **Recursive case:** print the number, then call `countUp(number + 1)`. **Progress:** the argument rises by 1 on each call, so it must reach 6. Prediction: 1 through 5, then `Count-up complete`.

```java
public class MissionCountUp {
    public static void countUp(int number) {
        if (number > 5) {
            System.out.println("Count-up complete");
        } else {
            System.out.println(number);
            countUp(number + 1);
        }
    }

    public static void main(String[] args) {
        countUp(1);
    }
}

MissionCountUp.main(null);
```

## Homework Hack: Recursive Power

**Base case:** `exponent == 0` returns 1, because any number to the power 0 is 1 and returning 0 would make every answer 0. **Recursive case:** `base * power(base, exponent - 1)`. **Progress:** the exponent drops by 1 on each call, so a nonnegative exponent reaches 0.

```java
public class MissionPowerHomework {
    public static int power(int base, int exponent) {
        if (exponent == 0) {
            return 1;
        }
        return base * power(base, exponent - 1);
    }

    public static void main(String[] args) {
        System.out.println("2^3 = " + power(2, 3));
        System.out.println("3^2 = " + power(3, 2));
    }
}

MissionPowerHomework.main(null);
```

## Call-Stack Trace of `power(2, 3)`

| Step | Call | Waits for | Returns |
|---|---|---|---|
| 1 | `power(2, 3)` | `2 * power(2, 2)` | 2 × 4 = **8** |
| 2 | `power(2, 2)` | `2 * power(2, 1)` | 2 × 2 = **4** |
| 3 | `power(2, 1)` | `2 * power(2, 0)` | 2 × 1 = **2** |
| 4 | `power(2, 0)` | Nothing, this is the base case | **1** |

The calls stack up from step 1 to step 4, and then the return values come back from step 4 to step 1.

## Stretch: Negative Exponents

With a negative exponent, `exponent - 1` moves further away from 0, so `exponent == 0` is never true. The calls keep going until Java throws a `StackOverflowError`. A robust method needs to handle that input up front, either by rejecting it with an `IllegalArgumentException` or by returning a `double` and computing `1 / power(base, -exponent)`.
