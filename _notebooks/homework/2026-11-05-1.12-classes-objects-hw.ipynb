## Exercise 2


```java

class Dog {
    String name;
}

public class AliasDemo {
    public static void main(String[] args) {
        Dog d3 = new Dog();
        d3.name = "Separate";
        Dog d1 = new Dog();
        d1.name = "Kai";

        Dog d2 = d1;          // copies the reference, not the object
        d2.name = "Kairo";    // changes the one shared object

        System.out.println("d1.name: " + d1.name);  // Kairo
        System.out.println("d2.name: " + d2.name);  // Kairo
        System.out.println("Same object? " + (d1 == d2));
        System.out.println("d3 separate? " + (d1 != d3));
        System.out.println("d3.name: " + d3.name);  // true
    }
}
AliasDemo.main(null);
```

## Exercise 3


```java

class Dog {
    String name;
}

public class SeparateObjects {
    public static void main(String[] args) {
        Dog e1 = new Dog();
        e1.name = "Changed"; // prediction: e2 stays Bo

        Dog e2 = new Dog();   // a second, separate object
        e2.name = "Bo";

        System.out.println("e1.name: " + e1.name);
        System.out.println("e2.name: " + e2.name);
        System.out.println("Same object? " + (e1 == e2));  // false, even though names match
    }
}
SeparateObjects.main(null);
```

## Exercise 4


```java

class Dog {
    String name;
}

public class ObjectDemo {
    public static void main(String[] args) {
        Dog x = new Dog();
        x.name = "Rex";
        System.out.println(x); // Object.toString() is used automatically.

        System.out.println(x.getClass().getName());   // Dog
        System.out.println(x instanceof Object);       // true

        // Every object inherits toString(), even without overriding it
        System.out.println(x.toString());               // Dog@<hash>, unless Dog overrides toString
    }
}
ObjectDemo.main(null);
```

## Exercise 6


```java
class Author {
    String name;
    void printName() { System.out.println(name); }
}
public class HomeworkClassesObjects {
    static String predictAliased = "Ada, Ada; same object true";
    static String predictSeparate = "Mira, Mira; same object false";
    public static void main(String[] args) {
        System.out.println("Predictions: aliased=" + predictAliased + ", separate=" + predictSeparate);
        Author a1 = new Author();
        Author a2 = a1;
        a2.name = "Ada";
        a1.printName();
        a2.printName();
        System.out.println("Aliased same object: " + (a1 == a2));
        Author b1 = new Author();
        Author b2 = new Author();
        b1.name = "Mira";
        b2.name = "Mira";
        b1.printName();
        b2.printName();
        System.out.println("Separate same object: " + (b1 == b2));
        // getClass() is inherited from Object. Prediction: class Author.
        System.out.println("Inherited getClass(): " + a1.getClass());
        boolean pass = a1 == a2 && a1.name.equals("Ada") && b1 != b2;
        System.out.println("Aliasing checks: " + pass);
        if (!pass) throw new AssertionError("Object identities do not match predictions");
    }
}
HomeworkClassesObjects.main(null);
```
