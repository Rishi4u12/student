## Homework Hack: Aliasing

Prediction: the aliased pair prints the same name twice because `a1` and `a2` refer to one object. The separate pair has equal names but `b1 == b2` is `false`, because `==` compares references and they are two objects.

```java
class Author {
    String name;

    void printName() {
        System.out.println("Author: " + name);
    }
}

public class HomeworkClassesObjects {
    static String predictAliased = "Octavia, Octavia";
    static String predictSeparate = "false";

    public static void main(String[] args) {
        System.out.println("Predictions: aliased=" + predictAliased + ", separate=" + predictSeparate);

        Author a1 = new Author();
        Author a2 = a1;
        a2.name = "Octavia";
        a1.printName();
        a2.printName();

        Author b1 = new Author();
        Author b2 = new Author();
        b1.name = "Ursula";
        b2.name = "Ursula";
        System.out.println("b1 == b2: " + (b1 == b2));

        // getClass() is inherited from Object, since Author doesn't define it.
        System.out.println("Class: " + a1.getClass().getSimpleName());
    }
}

HomeworkClassesObjects.main(null);
```

## Homework Page: Imaginary Students

My own instance variable is `favoriteSubject`. The second scenario is at the end of `main`: `nickname` is another reference to the same student, so it prints the same data and no new object is created.

```java
public class Student {
    String name;
    int grade;
    int pets;
    int siblings;
    String favoriteSubject;

    public Student(String name, int grade, int pets, int siblings, String favoriteSubject) {
        this.name = name;
        this.grade = grade;
        this.pets = pets;
        this.siblings = siblings;
        this.favoriteSubject = favoriteSubject;
    }

    void printInfo() {
        System.out.println(name + ", grade " + grade + ", " + pets + " pets, " + siblings
                + " siblings, likes " + favoriteSubject);
    }

    public static void main(String[] args) {
        Student first = new Student("Maya Lin", 11, 2, 1, "Physics");
        Student second = new Student("Theo Park", 12, 0, 3, "History");
        Student third = new Student("Priya Rao", 10, 1, 0, "Math");
        first.printInfo();
        second.printInfo();
        third.printInfo();

        Student nickname = second;
        System.out.println("Nickname reference:");
        nickname.printInfo();
        System.out.println("Same object: " + (nickname == second));
    }
}

Student.main(null);
```
