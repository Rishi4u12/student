## Popcorn Hack: Multiples of 5

Stepping by 5 visits only the values that print, so no `if` is needed. Prediction: 5 through 50, with a count of 10.

```java
public class Multiples {
    public static void main(String[] args) {
        int printed = 0;

        for (int i = 5; i <= 50; i += 5) {
            System.out.print(i + " ");
            printed++;
        }

        System.out.println();
        System.out.println("Printed " + printed + " numbers");   // expected: 10
    }
}

Multiples.main(null);
```

## Homework Hack: Factorial

`result` starts at 1 because starting at 0 would make every product 0. Prediction: 5! = 120, 1! = 1, and 7! = 5040.

```java
public class Factorial {
    public static void printFactorial(int n) {
        int result = 1;
        for (int i = 1; i <= n; i++) {
            result *= i;
        }
        System.out.println(n + "! = " + result);
    }

    public static void main(String[] args) {
        printFactorial(5);
        printFactorial(1);
        printFactorial(7);
    }
}

Factorial.main(null);
```
