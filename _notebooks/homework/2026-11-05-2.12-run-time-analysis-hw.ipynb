## Popcorn Hack: Faster Loop

Stepping by 5 skips the 108 values that never print, and removes the need for the `if`. Prediction: the same numbers, 0 through 130, in 27 passes.

```java
public class FasterLoop {
    public static void main(String[] args) {
        int passes = 0;

        for (int k = 0; k < 135; k += 5) {
            passes++;
            System.out.print(k + " ");
        }

        System.out.println();
        System.out.println("Passes: " + passes);   // goal: 27
    }
}

FasterLoop.main(null);
```

## Homework Hack: Predict the Counts

- **Loop 1 prediction: 4.** `i` takes 3, 8, 13, and 18.
- **Loop 2 prediction: 6.** The inner loop runs 0 + 1 + 2 + 3 times.
- **Loop 3 prediction: 3.** `n` goes 64 → 16 → 4 → 1.

```java
public class PredictCounts {
    public static void main(String[] args) {
        int count1 = 0;
        int count2 = 0;
        int count3 = 0;

        for (int i = 3; i <= 18; i += 5) {
            System.out.println("Loop 1: " + i);
            count1++;
        }

        for (int i = 0; i < 4; i++) {
            for (int j = 0; j < i; j++) {
                System.out.println("Loop 2: " + i + "," + j);
                count2++;
            }
        }

        int n = 64;
        while (n > 1) {
            n = n / 4;
            System.out.println("Loop 3: " + n);
            count3++;
        }

        System.out.println("Counts: " + count1 + ", " + count2 + ", " + count3);
    }
}

PredictCounts.main(null);
```

**12 times.** The outer loop runs 6 times. The inner loop's `j` takes 0 and 3, which is 2 passes, and 6 × 2 = 12.
