## Popcorn Hack: Attendant Shift

Prediction: unfilled spots in the `int` array print `0`, and the unfilled spot in the `String` array prints `null`. `new` fills an array with default values.

```java
public class AttendantShift {
    public static void main(String[] args) {
        int[] spots = new int[5];
        spots[1] = 214;
        spots[3] = 187;
        spots[4] = 305;
        for (int i = 0; i < spots.length; i++) {
            System.out.println("Spot " + i + ": " + spots[i]);
        }

        String[] drivers = new String[3];
        drivers[0] = "Lena";
        drivers[2] = "Marco";
        int i = 0;
        while (i < drivers.length) {
            System.out.println("Driver " + i + ": " + drivers[i]);
            i++;
        }
    }
}

AttendantShift.main(null);
```

## Homework Hack: Longest-Parked Car

The service charge needs an indexed loop. An enhanced `for` variable is only a copy, so changing it wouldn't change the array.

```java
public class LongestParked {
    public static void main(String[] args) {
        final String GARAGE_NAME = "Harbor Street Garage";
        int[] tickets = {1042, 987, 1103, 955, 1020};
        double[] fees = {12.5, 18.0, 6.75, 22.25, 9.0};

        int earliest = tickets[0];
        for (int i = 1; i < tickets.length; i++) {
            if (tickets[i] < earliest) {
                earliest = tickets[i];
            }
        }

        double highestFee = fees[0];
        for (double fee : fees) {
            if (fee > highestFee) {
                highestFee = fee;
            }
        }

        for (int i = 0; i < fees.length; i++) {
            fees[i] += 1.00;
        }

        System.out.println(GARAGE_NAME);
        System.out.println("Earliest ticket: " + earliest);
        System.out.println("Highest fee: " + highestFee);
        System.out.println("Fee at spot 0 after service charge: " + fees[0]);
    }
}

LongestParked.main(null);
```
