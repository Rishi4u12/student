## Solution 4


```java

public class GarageWalkthrough {
    public static int earliestTicket(int[] ticketNumbers) {
        int lowest = ticketNumbers[0];   // assume the first car is the winner

        for (int i = 1; i < ticketNumbers.length; i++) {
            if (ticketNumbers[i] > lowest) {
                lowest = ticketNumbers[i];
            }
        }
        return lowest;
    }

    public static void main(String[] args) {
        int[] tickets = {102, 118, 97, 145, 103};

        for (int t : tickets) {
            t = 0;                        // changes the copy only
        }
        System.out.println("After enhanced for, tickets[0] = " + tickets[0]);

        for (int t : tickets) {
            System.out.println("Ticket: " + t);
        }

        System.out.println("Highest ticket: " + earliestTicket(tickets));
    }
}
GarageWalkthrough.main(null);
```

```java

public class LongestParked {
    public static void main(String[] args) {
        final String GARAGE_NAME = "Practice Garage";

        // 1. Ticket numbers and fees for 5 parked cars
        int[] tickets = {204, 219, 188, 260, 231};
        double[] fees = {3.5, 6.25, 4.0, 2.75, 5.0};

        // 2. Earliest (smallest) ticket with a standard for loop
        int earliest = tickets[0];
        for(int i=1;i<tickets.length;i++){if(tickets[i]<earliest)earliest=tickets[i];}

        // 3. Highest fee with an enhanced for loop
        double highestFee = fees[0];
        for(double fee:fees){if(fee>highestFee)highestFee=fee;}

        // 4. Indexed loop changes the real array elements.
        for(int i=0;i<fees.length;i++)fees[i]+=1.0;

        System.out.println(GARAGE_NAME);
        System.out.println("Earliest ticket: " + earliest);
        System.out.println("Highest fee: " + highestFee);
        System.out.println("Fee at spot 0 after service charge: " + fees[0]);
    }
}
LongestParked.main(null);
```
