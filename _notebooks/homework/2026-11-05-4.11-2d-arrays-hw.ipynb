## Reflection Questions

1. **Why is a 2D array better than a 1D array here?** It keeps each subgroup together. A 1D array would lose which members belong together unless you did extra math on the index.
2. **Which index represents the row?** The first index, which selects the subgroup.
3. **Which index represents the column?** The second index, which selects a member within the subgroup.
