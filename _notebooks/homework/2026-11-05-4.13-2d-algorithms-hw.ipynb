## Chapter 1: The Full Sweep

Row-major traversal. `col` is 0-indexed, so the printed seat number is `col + 1`.

```java
public class SeatingChartPrinter {
    public static void printSeatingChart(int[][] chart, String[] podNames) {
        for (int row = 0; row < chart.length; row++) {
            for (int col = 0; col < chart[row].length; col++) {
                String seatValue = (chart[row][col] == -1) ? "EMPTY" : String.valueOf(chart[row][col]);
                System.out.println(podNames[row] + ", Seat " + (col + 1) + ": " + seatValue);
            }
        }
    }

    public static void main(String[] args) {
        printSeatingChart(SeatingData.scoreChart, SeatingData.podNames);
    }
}

SeatingChartPrinter.main(null);
```

## Chapter 2: Needle in Haystack

Returning from inside the nested loop stops the search at the first match. `{-1, -1}` means the score isn't in the grid.

```java
public class SeatSearch {
    public static int[] findSeat(int[][] chart, int targetScore) {
        for (int row = 0; row < chart.length; row++) {
            for (int col = 0; col < chart[row].length; col++) {
                if (chart[row][col] == targetScore) {
                    return new int[] {row, col};
                }
            }
        }
        return new int[] {-1, -1};
    }

    public static void main(String[] args) {
        int[] result = findSeat(SeatingData.scoreChart, 94);
        System.out.println("Found at: Pod " + result[0] + ", seat index " + result[1]);
    }
}

SeatSearch.main(null);
```

## Chapter 3: The Headcount

One traversal fills all the accumulators. The average divides by occupied seats only, is guarded against an empty room, and casts to `double` so it doesn't truncate.

```java
public class Headcount {
    public static double[] classSummary(int[][] chart) {
        int emptyCount = 0, occupiedCount = 0, totalScore = 0;
        for (int row = 0; row < chart.length; row++) {
            for (int col = 0; col < chart[row].length; col++) {
                if (chart[row][col] == -1) {
                    emptyCount++;
                } else {
                    occupiedCount++;
                    totalScore += chart[row][col];
                }
            }
        }
        double average = (occupiedCount == 0) ? 0.0 : (double) totalScore / occupiedCount;
        return new double[] {emptyCount, occupiedCount, average};
    }

    public static void main(String[] args) {
        double[] s = classSummary(SeatingData.scoreChart);
        System.out.println("Empty: " + s[0] + " | Occupied: " + s[1] + " | Avg: " + s[2]);
    }
}

Headcount.main(null);
```

## Capstone: SeatFinder

The champion starts at `-1`, so any real score beats it. Strict `>` keeps the first seat on a tie.

```java
public class TopScorer {
    public static int[] findTopScorer(int[][] chart) {
        int bestScore = -1, bestRow = -1, bestCol = -1;
        for (int row = 0; row < chart.length; row++) {
            for (int col = 0; col < chart[row].length; col++) {
                if (chart[row][col] != -1 && chart[row][col] > bestScore) {
                    bestScore = chart[row][col];
                    bestRow = row;
                    bestCol = col;
                }
            }
        }
        return new int[] {bestRow, bestCol, bestScore};
    }

    public static void main(String[] args) {
        int[] top = findTopScorer(SeatingData.scoreChart);
        System.out.println("Top: " + SeatingData.podNames[top[0]] + ", seat " + (top[1] + 1) + ", score " + top[2]);
    }
}

TopScorer.main(null);
```

## Edge Cases

```java
import java.util.Arrays;

public class EdgeCases {
    public static void main(String[] args) {
        int[][] allAbsent = { {-1, -1}, {-1, -1} };
        int[][] jagged = { {50}, {60, 70, 80}, {} };

        System.out.println("Score not in grid: " + Arrays.toString(SeatSearch.findSeat(SeatingData.scoreChart, 100)));
        System.out.println("Everyone absent: " + Arrays.toString(Headcount.classSummary(allAbsent)));
        System.out.println("No top scorer: " + Arrays.toString(TopScorer.findTopScorer(allAbsent)));
        System.out.println("Jagged top scorer: " + Arrays.toString(TopScorer.findTopScorer(jagged)));
    }
}

EdgeCases.main(null);
```
