## Temperature Data Processor Lab

`countHeatWaves` stops at `length - 1` because it reads `temps[i + 1]`. `reverseLog` swaps from both ends and stops when they meet, since continuing would swap everything back. The tests include an empty array and an odd-length array.

```java
import java.util.Arrays;

public class TemperatureAnalyzer {
    /**
     * Counts how many times two consecutive days both had
     * temperatures strictly greater than 90 degrees.
     */
    public static int countHeatWaves(int[] temps) {
        int count = 0;
        for (int i = 0; i < temps.length - 1; i++) {
            if (temps[i] > 90 && temps[i + 1] > 90) {
                count++;
            }
        }
        return count;
    }

    /**
     * Reverses the elements of the temps array in-place
     * using a while loop and a swap mechanism.
     */
    public static void reverseLog(int[] temps) {
        int left = 0;
        int right = temps.length - 1;
        while (left < right) {
            int temp = temps[left];
            temps[left] = temps[right];
            temps[right] = temp;
            left++;
            right--;
        }
    }

    public static void main(String[] args) {
        int[] week = {88, 91, 95, 90, 93, 94, 96};
        int[] empty = {};
        System.out.println("Heat waves: " + countHeatWaves(week));      // 91-95, 93-94, 94-96: expect 3
        System.out.println("Heat waves, empty: " + countHeatWaves(empty)); // expect 0

        reverseLog(week);
        System.out.println("Reversed: " + Arrays.toString(week));
        reverseLog(empty);
        System.out.println("Reversed, empty: " + Arrays.toString(empty));
    }
}

TemperatureAnalyzer.main(null);
```
