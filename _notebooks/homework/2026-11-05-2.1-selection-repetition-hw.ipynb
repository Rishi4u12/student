## Popcorn Hack: Vending Machine

Prediction, before running: 13 `Inserted` lines, because 130 / 10 is 13. Change is 0, so the `else` prints `Exact amount. No change.` The `while` is the repetition, and the `if` is the selection.

```java
public class PopcornMachine {
    public static void main(String[] args) {
        int price = 130;
        int total = 0;
        int coin = 10;
        int inserted = 0;

        while (total < price) {
            total = total + coin;
            inserted++;
            System.out.println("Inserted " + coin + " cents. Total: " + total);
        }
        System.out.println("Coins inserted: " + inserted);

        if (total > price) {
            System.out.println("Change: " + (total - price) + " cents");
        } else {
            System.out.println("Exact amount. No change.");
        }
    }
}

PopcornMachine.main(null);
```

## Homework Part 1: Algorithm

**Algorithm: Fill a Water Bottle at a Slow Fountain**

1. Pick up the empty bottle.
2. Set the amount in the bottle to 0 ounces.
3. Walk to the fountain and press the button.
4. WHILE the bottle has less than 24 ounces, add 4 ounces and check the level again. *(repetition)*
5. Release the button and screw on the cap.
6. IF the bottle has more than 24 ounces, pour out the extra. OTHERWISE, leave it as is. *(selection)*
7. Put the bottle in your bag.

## Homework Part 2: Java

The `while` loop repeats filling until the goal is reached. The `if` chooses what happens to the extra water, and it takes the `else` path here because 24 is an exact multiple of 4.

```java
public class MyAlgorithm {
    public static void main(String[] args) {
        int ounces = 0;
        int goal = 24;

        System.out.println("Pick up the empty bottle");
        System.out.println("Press the fountain button");
        while (ounces < goal) {
            ounces += 4;
            System.out.println("Bottle now has " + ounces + " ounces");
        }
        System.out.println("Release the button and screw on the cap");

        if (ounces > goal) {
            System.out.println("Pour out " + (ounces - goal) + " ounces");
        } else {
            System.out.println("Exactly full, nothing to pour out");
        }
        System.out.println("Put the bottle in your bag");
    }
}

MyAlgorithm.main(null);
```
