## Popcorn Hack: Grade Evaluator

Only the first true condition runs, so each branch only has to check its lower bound. Prediction: 90 is an A, 75 a C, 61 a D, and 40 an F.

```java
public class GradeEvaluator {
    public static void evaluate(double myAverage) {
        System.out.println("Current average: " + myAverage);
        if (myAverage >= 90) {
            System.out.println("You have an A average.");
        } else if (myAverage >= 80) {
            System.out.println("You have a B average.");
        } else if (myAverage >= 70) {
            System.out.println("You have a C average.");
        } else if (myAverage >= 60) {
            System.out.println("You have a D average.");
        } else {
            System.out.println("You have an F average.");
        }
    }

    public static void main(String[] args) {
        evaluate(90);
        evaluate(75);
        evaluate(61);
        evaluate(40);
    }
}

GradeEvaluator.main(null);
```

## Homework Hack: Member Discount

The spending question sits inside the member check, so it's only asked for members. Prediction for all four combinations: member at 120 gets 20%, member at 80 gets 10%, and non-members get 0% either way.

```java
public class Discount {
    public static void printPrice(boolean isMember, double total) {
        double discount = 0;
        if (isMember) {
            if (total >= 100) {
                discount = 0.20;
            } else {
                discount = 0.10;
            }
        }
        System.out.println("Member " + isMember + ", spent " + total);
        System.out.println("Discount: " + (discount * 100) + "%");
        System.out.println("You pay: $" + (total - total * discount));
    }

    public static void main(String[] args) {
        printPrice(true, 120.0);
        printPrice(true, 80.0);
        printPrice(false, 120.0);
        printPrice(false, 80.0);
    }
}

Discount.main(null);
```
