```java
public class GroupTraversal {
    public static void main(String[] args) {
        String[][] groups = {
            {"Rebecca", "Cyrus", "Rishabh"},
            {"Member 4", "Member 5", "Member 6"},
            {"Member 7", "Member 8", "Member 9"}
        };

        // Member 5 is located at row 1, column 1.
        System.out.println(groups[1][1]);

        for (int row = 0; row < groups.length; row++) {
            for (int col = 0; col < groups[row].length; col++) {
                System.out.println(groups[row][col]);
            }
        }
    }
}

GroupTraversal.main(null);
```

## Reflection

1. **Why do we need nested loops?** A 2D array is an array of rows, and each row is its own array, so one loop has to pick the row and another has to walk through it.
2. **What does the outer loop control?** Which row is being visited.
3. **What does the inner loop control?** Which column within that row is being visited.
4. **Why is `.length` better than hard-coding?** It stays correct when the array's size changes, and it handles rows of different lengths.
