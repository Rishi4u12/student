## Popcorn Hack: Formatting Tools

```java
public class FormattingTools {
    public static void main(String[] args) {
        System.out.print("Unit ");                          // print: the next output stays on this line
        System.out.println("One");                          // println: the next output starts below
        System.out.println("Top row\nBottom row");          // \n: line break inside one string
        System.out.println("The sign said \"Open\"");       // \": quotation marks inside a string
        System.out.println("Path: D:\\Projects");           // \\: prints one backslash
        System.out.printf("Total: $%.2f\n", 12.5);          // printf: %.2f shows exactly 2 decimals
        // Menu.java: println gives each option its own line, print keeps the typed answer
        // beside "Choose an option: ", and + joins the numbers into the output sentences.
    }
}

FormattingTools.main(null);
```

## Part 2: Fix the Bug

`\U` and `\S` are not valid escape sequences. Each backslash needs to be written as `\\`.

```java
public class BackslashBug {
    public static void main(String[] args) {
        System.out.println("C:\\Users\\Student");
    }
}

BackslashBug.main(null);
```

## Part 5: Calculator Menu

```java
import java.util.Scanner;

public class CalculatorMenu {
    public static void main(String[] args) {
        // The runner has no keyboard, so the choice and numbers come from a String.
        Scanner sc = new Scanner("1\n10\n5\n");

        System.out.println("==== Calculator Menu ====");
        System.out.println("1. Add");
        System.out.println("2. Subtract");
        System.out.println("3. Multiply");
        System.out.println("4. Divide");

        System.out.print("Choose an option: ");
        int choice = sc.nextInt();
        System.out.print("Enter first number: ");
        int first = sc.nextInt();
        System.out.print("Enter second number: ");
        int second = sc.nextInt();

        int result = 0;
        if (choice == 1) {
            result = first + second;
        } else if (choice == 2) {
            result = first - second;
        } else if (choice == 3) {
            result = first * second;
        } else if (choice == 4) {
            result = first / second;
        }
        System.out.println("Result: " + result);
    }
}

CalculatorMenu.main(null);
```
