## Popcorn Hack: Discount Rule

The member check was backwards: `isMember == false` should simply be `isMember`. Parentheses make the grouping clear, since `&&` happens before `||`. Prediction: non-member student gets no discount, member student gets a discount, and a senior gets one regardless.

```java
public class DiscountRule {
    public static void check(int age, boolean isStudent, boolean isMember) {
        if (age >= 65 || (isStudent && isMember)) {
            System.out.println("Discount!");
        } else {
            System.out.println("No discount");
        }
    }

    public static void main(String[] args) {
        check(20, true, false);
        check(20, true, true);
        check(70, false, false);
    }
}

DiscountRule.main(null);
```

## Homework Hack: Leap Year

Prediction: 2024 and 2000 are leap years, while 1900 and 2023 are not. 1900 fails because it's divisible by 100 but not 400.

```java
public class LeapYear {
    public static void check(int year) {
        boolean isLeap = (year % 4 == 0 && year % 100 != 0) || year % 400 == 0;

        if (isLeap) {
            System.out.println(year + " is a leap year.");
        } else {
            System.out.println(year + " is not a leap year.");
        }
    }

    public static void main(String[] args) {
        check(2024);
        check(1900);
        check(2000);
        check(2023);
    }
}

LeapYear.main(null);
```
