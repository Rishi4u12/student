## Popcorn Hack 1: Extend the Dog Class

Added a second non-void method, `getAge()`, and a second `Dog`.

```java
public class Dog {
    public String Name;
    public int Age;

    public Dog(String name, int age) {
        Name = name;
        Age = age;
    }

    public String getName() {
        return Name;
    }

    public int getAge() {
        return Age;
    }

    public static void main(String[] args) {
        Dog d = new Dog("Larry", 4);
        Dog other = new Dog("Rosie", 7);
        System.out.println(d.getName() + " is " + d.getAge());
        System.out.println(other.getName() + " is " + other.getAge());
    }
}

Dog.main(null);
```

## Popcorn Hack 2: Find a Substring's Position

The last lines print the original `String` again to show that `toUpperCase` returned a new `String` and left it unchanged.

```java
public class SubstringFinder {
    public static int findPosition(String full, String target) {
        return full.indexOf(target);
    }

    public static void main(String[] args) {
        String phrase = "object instantiation";
        System.out.println("inst starts at: " + findPosition(phrase, "inst"));
        System.out.println("zebra starts at: " + findPosition(phrase, "zebra"));

        String upper = phrase.toUpperCase();
        System.out.println("Returned: " + upper);
        System.out.println("Original: " + phrase);
    }
}

SubstringFinder.main(null);
```

## Popcorn Hack 3: Math-Powered Guessing Game

The target comes from `Math.random()`, `Math.abs()` measures how close each guess is, and `Math.pow()` turns the distance into a score penalty. The computer guesses by moving halfway toward the target, so no keyboard is needed.

```java
public class NumberGuessingGame {
    public static void main(String[] args) {
        int target = (int) (Math.random() * 100) + 1;
        int guess = 50;
        int score = 10000;
        int tries = 0;

        while (guess != target) {
            tries++;
            int distance = Math.abs(target - guess);
            score -= (int) Math.pow(distance, 2);
            System.out.println("Guess " + guess + " is off by " + distance);
            guess += (target - guess) / 2;
            if (Math.abs(target - guess) == 1) {
                guess = target;
            }
        }
        System.out.println("Found " + target + " after " + tries + " misses. Score: " + Math.max(score, 0));
    }
}

NumberGuessingGame.main(null);
```
