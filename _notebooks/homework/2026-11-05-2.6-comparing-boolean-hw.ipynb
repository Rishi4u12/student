## Popcorn Hack: Remove Every `!`

De Morgan's Law turns `!(age >= 18 || !member)` into `!(age >= 18) && !!member`, which simplifies to `age < 18 && member`.

Prediction: `(15, true)` is the youth rate, and `(20, false)` and `(20, true)` are standard. The cell prints both versions to confirm they agree.

```java
public class Simplify {
    public static void check(int age, boolean member) {
        boolean original = !(age >= 18 || !member);
        boolean simplified = age < 18 && member;
        System.out.println(age + ", " + member + ": " + (simplified ? "Youth member rate" : "Standard rate")
                + " (matches original: " + (original == simplified) + ")");
    }

    public static void main(String[] args) {
        check(15, true);
        check(20, false);
        check(20, true);
    }
}

Simplify.main(null);
```

## Homework Hack: Password Check

`==` compares references, and two `new String` objects are different references even with the same text, so `.equals` is the fix. The length check uses `<` and needs no `!`. Prediction: `Welcome back!` and `Length OK`, since `open-sesame` has 11 characters.

```java
public class PasswordCheck {
    public static void main(String[] args) {
        String saved = new String("open-sesame");
        String typed = new String("open-sesame");

        if (typed.equals(saved)) {
            System.out.println("Welcome back!");
        } else {
            System.out.println("Wrong password.");
        }

        if (typed.length() < 8) {
            System.out.println("Too short");
        } else {
            System.out.println("Length OK");
        }
    }
}

PasswordCheck.main(null);
```
