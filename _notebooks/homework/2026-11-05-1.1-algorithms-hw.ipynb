## Popcorn Hack: Store Closing Algorithm

All 10 steps are kept and only reordered. The front door locks first so no new customers arrive, and the employee door locks last because it's how you leave.

```java
public class StoreClosing {
    public static void main(String[] args) {
        System.out.println("Lock the front door after the last customer leaves");
        System.out.println("Close out cash registers and put the money somewhere safe");
        System.out.println("Straighten up the items on the shelves");
        System.out.println("Count inventory");
        System.out.println("Make a list of items that need to be restocked");
        System.out.println("Take out the trash");
        System.out.println("Mop the floor");
        System.out.println("Turn off all of the lights");
        System.out.println("Turn on the security system and cameras");
        System.out.println("Exit through the employee door and lock it");
    }
}

StoreClosing.main(null);
```

## Homework Part A: Making a Pizza From Scratch

1. Mix flour, yeast, salt, olive oil, and warm water into a dough.
2. Knead the dough for 8 minutes until it is smooth.
3. Cover the dough and let it rise for 1 hour.
4. Preheat the oven to 475°F.
5. Stretch the dough into a 12-inch circle on a floured pan.
6. Spread sauce over the dough, leaving a 1-inch border.
7. Add cheese and toppings evenly.
8. Bake for 12 minutes until the crust is golden.
9. Rest the pizza for 2 minutes, then slice and serve.

## Homework Part C: Letter Grade Calculator

```java
public class LetterGradeCalculator {
    public static String calculateGrade(int score1, int score2, int score3) {
        int total = score1 + score2 + score3;
        double average = total / 3.0;  // 3.0 keeps the decimals, so 89.9 stays a B
        if (average >= 90) {
            return "A";
        } else if (average >= 80) {
            return "B";
        } else if (average >= 70) {
            return "C";
        } else if (average >= 60) {
            return "D";
        } else {
            return "F";
        }
    }

    public static void main(String[] args) {
        System.out.println("Test 1: " + calculateGrade(95, 92, 88));  // Should be 'A'
        System.out.println("Test 2: " + calculateGrade(85, 80, 82));  // Should be 'B'
        System.out.println("Test 3: " + calculateGrade(75, 70, 72));  // Should be 'C'
        System.out.println("Test 4: " + calculateGrade(65, 60, 62));  // Should be 'D'
        System.out.println("Test 5: " + calculateGrade(55, 50, 52));  // Should be 'F'
    }
}

LetterGradeCalculator.main(null);
```
