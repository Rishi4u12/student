```java

public class LetterGradeCalculator {
    /**
     * Calculate letter grade from three test scores
     *
     * @param score1 first test score
     * @param score2 second test score
     * @param score3 third test score
     * @return the letter grade
     */
    public static String calculateGrade(int score1, int score2, int score3) {
        // Step 1 and 2: Add the three scores together, then calculate the average
        double average = (score1 + score2 + score3) / 3.0;

        // Step 3: Determine the letter grade using if-else if-else
        if (average >= 90) {
            return "A";
        } else if (average >= 80) {
            return "B";
        } else if (average >= 70) {
            return "C";
        } else if (average >= 60) {
            return "D";
        } else {
            return "F";
        }
    }

    public static void main(String[] args) {
        // Test your function!
        System.out.println("A".equals(calculateGrade(95, 92, 88)));  // Test 1: should be A
        System.out.println("B".equals(calculateGrade(85, 80, 82)));  // Test 2: should be B
        System.out.println("C".equals(calculateGrade(75, 70, 72)));  // Test 3: should be C
        System.out.println("D".equals(calculateGrade(65, 60, 62)));  // Test 4: should be D
        System.out.println("F".equals(calculateGrade(55, 50, 52)));  // Test 5: should be F
        System.out.println("A".equals(calculateGrade(90, 90, 90)));  // Test 6: exactly 90 should be A
    }
}

LetterGradeCalculator.main(null);
```
