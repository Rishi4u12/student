## Homework Hack: Fix the Calls

The original `Game.showPlayer()` failed because `showPlayer()` is an instance method that reads `player`. It needs a `Game` object to know whose player to show, and a class-name call doesn't provide one.

```java
public class Game {
    String player;

    public Game(String player) {
        this.player = player;
    }

    public void showPlayer() {
        System.out.println("Player: " + player);
    }

    public static void rules() {
        System.out.println("Each player gets one turn.");
    }

    public static void main(String[] args) {
        Game.rules();
        Game game = new Game("Ada");
        game.showPlayer();
    }
}

Game.main(null);
```

## Homework: Ultimate Battle

The assignment lists `attack()` and `strongerFighter()` with no parameters, so each fighter stores its opponent and the class stores the two fighters. My own additions are the `armor` variable, the `heal()` instance method, and the `printFact()` class method.

```java
public class BattleRobot {
    String name;
    int power;
    int health;
    int armor;
    BattleRobot opponent;

    static double fightDuration = 0.0;
    static BattleRobot first;
    static BattleRobot second;

    public BattleRobot(String name, int power, int health, int armor) {
        this.name = name;
        this.power = power;
        this.health = health;
        this.armor = armor;
    }

    void attack() {
        int damage = Math.max(power - opponent.armor, 0);
        opponent.health -= damage;
        System.out.println(name + " hits " + opponent.name + " for " + damage);
    }

    void printStatus() {
        System.out.println(name + ": health " + health);
    }

    void heal() {
        health += 5;
        System.out.println(name + " heals to " + health);
    }

    static int strongerFighter() {
        return first.power >= second.power ? 1 : 2;
    }

    static void beginBattle() {
        System.out.println("Battle start! Fighter " + strongerFighter() + " has more power.");
        while (first.health > 0 && second.health > 0) {
            first.attack();
            if (second.health > 0) {
                second.attack();
            }
            fightDuration += 1.5;
        }
        String winner = first.health > 0 ? first.name : second.name;
        System.out.println(winner + " wins after " + fightDuration + " seconds");
    }

    static void printFact() {
        System.out.println("Fact: every robot shares one fightDuration because it is static.");
    }

    public static void main(String[] args) {
        first = new BattleRobot("Bolt", 12, 30, 2);
        second = new BattleRobot("Tank", 9, 40, 4);
        first.opponent = second;
        second.opponent = first;

        BattleRobot.printFact();
        first.attack();
        second.heal();
        first.printStatus();
        second.printStatus();
        BattleRobot.beginBattle();
        first.printStatus();
        second.printStatus();
    }
}

BattleRobot.main(null);
```
