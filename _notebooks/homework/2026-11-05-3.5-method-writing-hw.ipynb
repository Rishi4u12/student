## Homework Hack 1: Returning Method

Returns `int` and takes two `int` parameters.

```java
public class PerimeterHack {
    public int perimeter(int length, int width) {
        return 2 * length + 2 * width;
    }

    public static void main(String[] args) {
        PerimeterHack hack = new PerimeterHack();
        System.out.println(hack.perimeter(8, 5)); // expected: 26
    }
}

PerimeterHack.main(null);
```

## Homework Hack 2: `void` Method

Prints directly, so it has nothing to return.

```java
public class DescriptionHack {
    public void printRectangle(int length, int width) {
        System.out.println("Rectangle: " + length + " by " + width);
    }

    public static void main(String[] args) {
        DescriptionHack hack = new DescriptionHack();
        hack.printRectangle(8, 5);
    }
}

DescriptionHack.main(null);
```

## Homework Hack 3: Return a Decision

The comparison already produces a `boolean`, so it can be returned directly.

```java
public class SquareHack {
    public boolean isSquare(int length, int width) {
        return length == width;
    }

    public static void main(String[] args) {
        SquareHack hack = new SquareHack();
        System.out.println(hack.isSquare(4, 4));  // expected: true
        System.out.println(!hack.isSquare(8, 5)); // expected: true
    }
}

SquareHack.main(null);
```
