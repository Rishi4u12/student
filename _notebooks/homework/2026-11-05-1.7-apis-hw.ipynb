## Popcorn Hack 2: Attributes and Behaviors

Attributes: `title`, `author`, `pages`. Behaviors: the constructor, `displayInfo()`, and `isLong()`.

```java
public class Book {
    private String title;
    private String author;
    private int pages;

    public Book(String title, String author, int pages) {
        this.title = title;
        this.author = author;
        this.pages = pages;
    }

    public void displayInfo() {
        System.out.println(title + " by " + author + ", " + pages + " pages");
    }

    public boolean isLong() {
        return pages > 300;
    }

    public static void main(String[] args) {
        Book myBook = new Book("Java Basics", "John Doe", 350);
        myBook.displayInfo();
        System.out.println("Long book: " + myBook.isLong());
    }
}

Book.main(null);
```

## Homework Hack: `java.util.HashMap`

Quoted method summaries are from the [Java SE `HashMap` documentation](https://docs.oracle.com/en/java/javase/21/docs/api/java.base/java/util/HashMap.html). All of these are behaviors, because `HashMap` exposes its data only through methods.

```java
import java.util.HashMap;

public class HashMapDemo {
    public static void main(String[] args) {
        HashMap<String, Integer> scores = new HashMap<String, Integer>();

        // Behavior. Doc for put: "Associates the specified value with the specified key in this map."
        scores.put("Ada", 95);
        scores.put("Linus", 88);

        // Behavior. Doc for get: "Returns the value to which the specified key is mapped, or null if this map contains no mapping for the key."
        System.out.println("Ada: " + scores.get("Ada"));

        // Behavior. Doc for containsKey: "Returns true if this map contains a mapping for the specified key."
        System.out.println("Has Grace: " + scores.containsKey("Grace"));

        // Behavior. Doc for size: "Returns the number of key-value mappings in this map."
        System.out.println("Entries: " + scores.size());
    }
}

HashMapDemo.main(null);
```
