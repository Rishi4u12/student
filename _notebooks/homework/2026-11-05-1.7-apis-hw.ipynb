## Popcorn Hack 1: Using Documentation

Prediction: Math.pow(3,4) gives 81.0 and Math.sqrt(64) gives 8.0. Adding red, blue, and green yields a list of size 3. Math belongs to java.lang, so it needs no import; ArrayList belongs to java.util, so it is imported. API references: [Math](https://docs.oracle.com/en/java/javase/17/docs/api/java.base/java/lang/Math.html#pow(double,double)) and [ArrayList](https://docs.oracle.com/en/java/javase/17/docs/api/java.base/java/util/ArrayList.html#add(E)). The standard methods perform the operations; no replacement algorithms are handwritten.

```java
import java.util.ArrayList;
public class CodeRunner7_PopcornHack1 {
    public static void main(String[] args) {
        double power = Math.pow(3, 4);
        double root = Math.sqrt(64);
        ArrayList<String> colors = new ArrayList<>();
        colors.add("red");
        colors.add("blue");
        colors.add("green");
        System.out.println("3^4: " + power);
        System.out.println("Square root of 64: " + root);
        System.out.println("Colors: " + colors);
        System.out.println("Size: " + colors.size());
        boolean pass = power == 81.0 && root == 8.0 && colors.size() == 3;
        System.out.println("Popcorn 1 check: " + pass);
        if (!pass) throw new AssertionError("API results differ from predictions");
    }
}

CodeRunner7_PopcornHack1.main(null);
```

## Popcorn Hack 2: Attributes and Behaviors

Book is a nested class inside the required CodeRunner8_PopcornHack2 runner. Its attributes are title, author, and pages; its displayInfo and isLong methods are behaviors. The constructor initializes all attributes. Prediction: the 350-page example returns true for isLong, while the 300-page boundary returns false because the condition is strictly greater than 300. Both objects are displayed and tested.

```java
public class CodeRunner8_PopcornHack2 {
    // Book attributes are the data each object HAS.
    static class Book {
        private String title;
        private String author;
        private int pages;
        // Constructor initializes all three attributes.
        public Book(String title, String author, int pages) {
            this.title = title;
            this.author = author;
            this.pages = pages;
        }
        // Behavior: displays all book attributes.
        public void displayInfo() {
            System.out.println(title + " by " + author + ", " + pages + " pages");
        }
        // Behavior: computes whether the page count exceeds 300.
        public boolean isLong() { return pages > 300; }
    }
    public static void main(String[] args) {
        // Fictional book values from the lesson's example.
        Book book = new Book("Java Basics", "John Doe", 350);
        book.displayInfo();
        boolean longBook = book.isLong();
        System.out.println("Long book: " + longBook);
        Book boundary = new Book("Boundary Example", "Example Author", 300);
        boundary.displayInfo();
        boolean atBoundary = boundary.isLong();
        System.out.println("Exactly 300 pages is long: " + atBoundary);
        boolean pass = longBook && !atBoundary;
        System.out.println("Popcorn 2 check: " + pass);
        if (!pass) throw new AssertionError("isLong boundary failed");
    }
}

CodeRunner8_PopcornHack2.main(null);
```

## Homework Hack: java.util.HashMap

HashMap is a java.util class not demonstrated in the lesson's examples. The official [Java SE 17 API documentation](https://docs.oracle.com/en/java/javase/17/docs/api/java.base/java/util/HashMap.html) specifies put, get, and containsKey. Each call below cites its exact signature line, method-detail heading, and stable URL anchor. The listed line numbers refer to the retrieved documentation text; the method anchors identify the corresponding official entries.

All three calls are behaviors because they perform operations through methods. No HashMap attribute is accessed directly. The scores variable stores the map reference, and the map internally stores its entries. Prediction: inserting Ada initially returns null, retrieving Ada gives 95, and checking the absent Grace key returns false.

Execution evidence: all three class bodies were run with local Java 17 and their actual stdout is stored below. The notebook cells invoke each main method. The school lesson has no interactive runners, its linked homework page returns 404, and no MCQ appears on the lesson page. This notebook completes the requirements stated on the main lesson page; no successful school submission is claimed.

```java
import java.util.HashMap;
public class HashMapDemo {
    public static void main(String[] args) {
        // Constructor creates a map; the variable scores holds its reference.
        HashMap<String, Integer> scores = new HashMap<>();
        // Behavior. Exact API signature: public V put(K key, V value)
        // Oracle Java SE 17 HashMap, Method Details > put (line 295).
        // https://docs.oracle.com/en/java/javase/17/docs/api/java.base/java/util/HashMap.html#put(K,V)
        Integer oldScore = scores.put("Ada", 95);
        // Behavior. Exact API signature: public V get(Object key)
        // Oracle Java SE 17 HashMap, Method Details > get (line 263).
        // https://docs.oracle.com/en/java/javase/17/docs/api/java.base/java/util/HashMap.html#get(java.lang.Object)
        Integer adaScore = scores.get("Ada");
        // Behavior. Exact API signature: public boolean containsKey(Object key)
        // Oracle Java SE 17 HashMap, Method Details > containsKey (line 282).
        // https://docs.oracle.com/en/java/javase/17/docs/api/java.base/java/util/HashMap.html#containsKey(java.lang.Object)
        boolean hasGrace = scores.containsKey("Grace");
        // These println calls are output behaviors, not direct attribute access.
        System.out.println("Previous Ada score: " + oldScore);
        System.out.println("Ada score: " + adaScore);
        System.out.println("Has Grace: " + hasGrace);
        boolean pass = oldScore == null && adaScore == 95 && !hasGrace;
        System.out.println("Homework check: " + (pass ? "PASS" : "FAIL"));
        if (!pass) throw new AssertionError("HashMap behavior did not match documentation");
    }
}

HashMapDemo.main(null);
```
