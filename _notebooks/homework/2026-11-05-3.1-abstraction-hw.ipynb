## Popcorn Hack: `calculateAverage`

The calculation now lives in a named method, so `main` only says what it wants. Prediction: `Average: 90.0`.

```java
public class PopcornAbstraction {
    public static double calculateAverage(double[] scores) {
        double total = 0;
        for (double score : scores) {
            total += score;
        }
        return total / scores.length;
    }

    public static void main(String[] args) {
        double[] scores = {80, 90, 100};
        double average = calculateAverage(scores);
        System.out.println("Average: " + average);
    }
}

PopcornAbstraction.main(null);
```

## Homework Hack: `PracticeGame`

`currentPoints` and `goalPoints` stay as state. `addPoints` is now the only place scoring happens, `hasReachedGoal` is the only place the goal is checked, and both turn methods call them instead of repeating the logic.

The lesson says to preserve the original behavior, but it only provides the starter and not the original class. These print messages are my own choice.

```java
public class PracticeGame {
    private int currentPoints = 0;
    private int goalPoints = 50;

    public void addPoints(int points) {
        currentPoints += points;
    }

    public boolean hasReachedGoal() {
        return currentPoints >= goalPoints;
    }

    private void reportProgress() {
        System.out.println("Points: " + currentPoints + ", goal reached: " + hasReachedGoal());
    }

    public void takeTurn(int earnedPoints) {
        addPoints(earnedPoints);
        reportProgress();
    }

    public void bonusTurn(int earnedPoints) {
        addPoints(earnedPoints);
        System.out.println("Bonus turn!");
        reportProgress();
    }

    public static void main(String[] args) {
        PracticeGame game = new PracticeGame();
        game.takeTurn(20);
        game.bonusTurn(35);
    }
}

PracticeGame.main(null);
```
