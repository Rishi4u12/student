## Popcorn Hack: Game Score With Compound Operators

Predicted trace: score 1000 + 250 = 1250; health 100 - 15 = 85; enemies 0 + 1 = 1; score 1250 * 2 = 2500; health 85 * 4 = 340, then 340 / 5 = 68. The final state is score 2500, health 68, and one enemy. Each compound assignment replaces the matching expanded form without changing this calculation. The post-increment increases the enemy counter by one. Every operation prints its new value.

```java
public class CodeRunner6_GameScoreCompound {
    public static void main(String[] args) {
        int playerScore = 1000;
        int playerHealth = 100;
        int enemiesDefeated = 0;
        playerScore += 250; // Add the enemy reward to the existing score.
        System.out.println("After reward, score: " + playerScore);
        playerHealth -= 15; // Subtract damage from current health.
        System.out.println("After damage, health: " + playerHealth);
        enemiesDefeated++; // Exactly one enemy was defeated; use post-increment.
        System.out.println("Enemies defeated: " + enemiesDefeated);
        playerScore *= 2; // Multiply the current score by the boss bonus.
        System.out.println("After boss bonus, score: " + playerScore);
        playerHealth *= 4; // Multiply before dividing to preserve the intended ratio.
        System.out.println("Health after multiplying by 4: " + playerHealth);
        playerHealth /= 5; // Complete the 4/5 scaling with integer division.
        System.out.println("Health after dividing by 5: " + playerHealth);
        System.out.println("Score: " + playerScore);
        System.out.println("Health: " + playerHealth);
        System.out.println("Enemies: " + enemiesDefeated);
        boolean correct = playerScore == 2500 && playerHealth == 68 && enemiesDefeated == 1;
        System.out.println("Popcorn check: " + correct);
        if (!correct) throw new AssertionError("Unexpected game state");
    }
}

CodeRunner6_GameScoreCompound.main(null);
```

## Homework Hack: Option A, Bank Account Manager

This uses fictional amounts from the lesson's scenario. Seven events change the account: deposit, rent, interest, fee, groceries/utilities, promotional multiplier, and savings transfer. Four different compound operators (+=, -=, *=, /=) model addition, removal, scaling, and division. Only post-form increment is used. Every compound assignment and increment prints the affected value immediately.

Predicted balances: 1000 -> 4000 -> 2800 -> 2814 -> 2809 -> 2659 -> 2685.59 -> 1342.795. Formatting the final balance to cents prints 1342.80. Interest totals 14.00, fees total 5.00, and all seven events are counted. Interest and fees are accumulated separately; only each new amount is applied to the balance, avoiding repeated application of prior totals. The final check compares the unrounded balance with a small tolerance because double arithmetic can contain rounding error. No external libraries are used.

Execution evidence: both class bodies were run with local Java 17; their actual stdout is stored below. The Java notebook cells call each class's main method. The lesson currently has no interactive runners, so no successful site execution is claimed.

```java
public class CodeRunner7_BankAccountManager {
    public static void main(String[] args) {
        // Fictional simulation values, not personal financial information.
        double balance = 1000.0;
        int transactions = 0;
        double totalInterest = 0.0;
        double totalFees = 0.0;
        System.out.printf("Starting balance: $%.2f%n", balance);
        // Event 1: += accumulates a deposit in the account.
        balance += 3000.0;
        System.out.printf("Deposit: balance $%.2f%n", balance);
        transactions++;
        System.out.println("Transactions: " + transactions);
        // Event 2: -= removes the rent payment.
        balance -= 1200.0;
        System.out.printf("Rent: balance $%.2f%n", balance);
        transactions++;
        System.out.println("Transactions: " + transactions);
        // Event 3: accumulate interest separately, then add this month's interest.
        double monthlyInterest = balance * 0.005;
        totalInterest += monthlyInterest;
        System.out.printf("Total interest: $%.2f%n", totalInterest);
        balance += monthlyInterest;
        System.out.printf("Interest credited: balance $%.2f%n", balance);
        transactions++;
        System.out.println("Transactions: " + transactions);
        // Event 4: accumulate fees, but deduct only the new fee.
        totalFees += 5.0;
        System.out.printf("Total fees: $%.2f%n", totalFees);
        balance -= 5.0;
        System.out.printf("Maintenance fee: balance $%.2f%n", balance);
        transactions++;
        System.out.println("Transactions: " + transactions);
        // Event 5: -= models the groceries and utilities payment.
        balance -= 150.0;
        System.out.printf("Groceries and utilities: balance $%.2f%n", balance);
        transactions++;
        System.out.println("Transactions: " + transactions);
        // Event 6: *= applies a promotional one-percent balance increase.
        balance *= 1.01;
        System.out.printf("Promotional multiplier: balance $%.2f%n", balance);
        transactions++;
        System.out.println("Transactions: " + transactions);
        // Event 7: /= retains half after transferring the other half to savings.
        balance /= 2.0;
        System.out.printf("Savings transfer: balance $%.2f%n", balance);
        transactions++;
        System.out.println("Transactions: " + transactions);
        System.out.printf("Final balance: $%.2f; interest: $%.2f; fees: $%.2f; transactions: %d%n",
            balance, totalInterest, totalFees, transactions);
        boolean correct = Math.abs(balance - 1342.795) < 0.000001
            && totalInterest == 14.0 && totalFees == 5.0 && transactions == 7;
        System.out.println("Homework check: " + (correct ? "PASS" : "FAIL"));
        if (!correct) throw new AssertionError("Unexpected bank state");
    }
}

CodeRunner7_BankAccountManager.main(null);
```
