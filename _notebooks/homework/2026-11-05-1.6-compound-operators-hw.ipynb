## Popcorn Hack: Game Score With Compound Operators

```java
public class GameScoreCompound {
    public static void main(String[] args) {
        int playerScore = 1000;
        int playerHealth = 100;
        int enemiesDefeated = 0;

        playerScore += 250;   // enemy worth 250 points
        System.out.println("Score: " + playerScore);
        playerHealth -= 15;   // took 15 damage
        System.out.println("Health: " + playerHealth);
        enemiesDefeated++;    // one more enemy down
        System.out.println("Enemies: " + enemiesDefeated);
        playerScore *= 2;     // boss battle doubles the score
        System.out.println("Score: " + playerScore);
        playerHealth *= 4;    // multiply first so the int division below keeps precision
        System.out.println("Health: " + playerHealth);
        playerHealth /= 5;    // finishes keeping 80% of health
        System.out.println("Health: " + playerHealth);
    }
}

GameScoreCompound.main(null);
```

## Homework Hack: Option A, Bank Account Manager

Each operator matches the kind of change it models, and the value prints after every operation.

```java
public class BankAccountManager {
    public static void main(String[] args) {
        double balance = 500.0;
        int transactions = 0;
        double totalFees = 0.0;

        balance += 1200.0;       // += a paycheck adds to the balance
        transactions++;          // ++ exactly one more transaction
        System.out.println("Paycheck: balance $" + balance + ", transactions " + transactions);

        balance -= 650.0;        // -= rent removes money
        transactions++;
        System.out.println("Rent: balance $" + balance + ", transactions " + transactions);

        balance *= 1.01;         // *= interest grows the balance by a percentage
        System.out.printf("Interest: balance $%.2f\n", balance);

        totalFees += 12.5;       // += fees accumulate in their own total
        balance -= 12.5;
        transactions++;
        System.out.printf("Fee: balance $%.2f, total fees $%.2f, transactions %d\n", balance, totalFees, transactions);

        balance /= 2;            // /= half the money moves to savings
        transactions++;
        System.out.printf("Savings transfer: balance $%.2f, transactions %d\n", balance, transactions);

        int leftoverCents = (int) (balance * 100);
        leftoverCents %= 100;    // %= keeps only the cents
        System.out.println("Cents past the dollar: " + leftoverCents);
    }
}

BankAccountManager.main(null);
```
