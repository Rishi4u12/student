## Popcorn Hack: Locker

Each fix keeps the variable names and uses `this` to reach the field.

```java
public class Locker {
    private String owner;
    private int items;

    public Locker(String owner) {
        // The broken line touched only the parameter owner, assigning it to itself.
        this.owner = owner;
        items = 0;
    }

    public void store(int count) {
        // The broken version declared a new local variable named items, which hid the field.
        for (int i = 0; i < count; i++) {
            int added = 1;
            this.items += added;
        }
        // System.out.println(added);
        // This can never compile: added was declared inside the loop, so its scope ended there.
    }

    public String getOwner() { return owner; }
    public int getItems() { return items; }

    public static void main(String[] args) {
        Locker l = new Locker("Maya");
        l.store(3);
        System.out.println(l.getOwner() + " " + l.getItems());
    }
}

Locker.main(null);
```

## Homework Hack: Player and Game

The `p.score = 9999` line stays commented out because it won't compile: javac reports `score has private access in Player`. That's the "both ways" comparison from the task: direct access is blocked, while `addPoints` works and can reject bad values.

```java
public class Player {
    private String name;
    private int score;

    public Player(String name, int score) {
        this.name = name;
        this.score = score;
    }

    public int getScore() {
        return score;
    }

    public boolean addPoints(int amount) {
        if (amount > 0) {
            score += amount;
            return true;
        }
        return false;
    }
}

public class Game {
    public static void main(String[] args) {
        Player p = new Player("Ravi", 25);
        System.out.println("start: " + p.getScore());              // 25
        System.out.println("add 5: " + p.addPoints(5));            // true
        System.out.println("add -50: " + p.addPoints(-50));        // false
        System.out.println("end: " + p.getScore());                // 30
        // p.score = 9999;   // error: score has private access in Player
    }
}

Game.main(null);
```
