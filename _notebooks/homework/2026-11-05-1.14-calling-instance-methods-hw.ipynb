## Popcorn Hack 1: Dog With a Breed

```java
public class Dog {
    private String name;
    private String breed;

    public Dog(String name, String breed) {
        this.name = name;
        this.breed = breed;
    }

    public void bark() {
        System.out.println(name + " says: Woof!");
    }

    public void describe() {
        System.out.println("My dog " + name + " is a " + breed + "!");
    }

    public static void main(String[] args) {
        Dog dog = new Dog("Fido", "Beagle");
        dog.bark();
        dog.describe();
    }
}

Dog.main(null);
```

## Popcorn Hack 2: Counter

```java
public class Counter {
    private int count;

    public void add(int x) { count += x; }
    public void subtract(int x) { count -= x; }
    public void multiply(int x) { count *= x; }
    public void divide(int x) { count /= x; }
    public int getCount() { return count; }

    public static void main(String[] args) {
        Counter c = new Counter();
        c.add(5);
        System.out.println("After add: " + c.getCount());
        c.subtract(2);
        System.out.println("After subtract: " + c.getCount());
        c.multiply(8);
        System.out.println("After multiply: " + c.getCount());
        c.divide(4);
        System.out.println("After divide: " + c.getCount());
    }
}

Counter.main(null);
```

## Homework Hack: Student Grade Tracker

```java
public class Student {
    private String name;
    private int totalPoints;
    private int numAssignments;

    public Student(String name) {
        this.name = name;
        totalPoints = 0;
        numAssignments = 0;
        System.out.println("Creating student: " + name);
        System.out.println("Student created successfully!");
    }

    // Adds one graded assignment to this student's totals.
    public void addGrade(int points) {
        totalPoints += points;
        numAssignments++;
        System.out.println("Adding grade: " + points + " points");
    }

    // Returns 0.0 until a grade exists, to avoid dividing by zero.
    public double getAverage() {
        if (numAssignments == 0) {
            return 0.0;
        }
        double average = (double) totalPoints / numAssignments;
        return Math.round(average * 100) / 100.0;
    }

    public String getLetterGrade() {
        double average = getAverage();
        if (average >= 90) return "A";
        if (average >= 80) return "B";
        if (average >= 70) return "C";
        if (average >= 60) return "D";
        return "F";
    }

    public void printReport() {
        System.out.println("Student Name: " + name);
        System.out.println("Total Points: " + totalPoints);
        System.out.println("Assignments Completed: " + numAssignments);
        System.out.println("Current Average: " + getAverage());
        System.out.println("Letter Grade: " + getLetterGrade());
        System.out.println("Status: " + (getAverage() >= 90 ? "Excellent work!" : "Keep working hard!"));
    }

    public String getSummary() {
        return name + " - Average: " + getAverage() + " (" + getLetterGrade() + ")";
    }

    public static void main(String[] args) {
        System.out.println("=== Student Grade Tracker System ===\n");

        Student nora = new Student("Nora Lee");
        System.out.println("\n--- Adding Grades for Nora ---");
        nora.addGrade(96);
        nora.addGrade(91);
        nora.addGrade(87);
        System.out.println("\n--- Nora's Grade Report ---");
        nora.printReport();

        System.out.println("\n========================================\n");

        Student omar = new Student("Omar Haddad");
        System.out.println("\n--- Adding Grades for Omar ---");
        omar.addGrade(72);
        omar.addGrade(81);
        omar.addGrade(70);
        omar.addGrade(76);
        System.out.println("\n--- Omar's Grade Report ---");
        omar.printReport();

        System.out.println("\n========================================\n");

        System.out.println("Final Summary:");
        System.out.println(nora.getSummary());
        System.out.println(omar.getSummary());
    }
}

Student.main(null);
```
