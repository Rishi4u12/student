## Popcorn Hack: Vowel Counter

`substring(i, i + 1)` gives a one-character `String`, so it needs `.equals` rather than `==`. Prediction: `iterate` has 4 vowels: i, e, a, e.

```java
public class VowelCounter {
    public static void main(String[] args) {
        String word = "iterate";
        int vowels = 0;

        for (int i = 0; i < word.length(); i++) {
            String letter = word.substring(i, i + 1);
            if (letter.equals("a") || letter.equals("e") || letter.equals("i")
                    || letter.equals("o") || letter.equals("u")) {
                vowels++;
            }
        }

        System.out.println("Vowels: " + vowels);   // expected: 4
    }
}

VowelCounter.main(null);
```

## Homework Hack: Palindrome

Each new letter goes in front of what's already built, which reverses the order. `.equals` compares the text. Prediction: `racecar` and `level` are true, and `java` is false.

```java
public class Palindrome {
    public static boolean isPalindrome(String s) {
        String rev = "";
        for (int i = 0; i < s.length(); i++) {
            rev = s.substring(i, i + 1) + rev;
        }
        return rev.equals(s);
    }

    public static void main(String[] args) {
        System.out.println("racecar: " + isPalindrome("racecar"));   // true
        System.out.println("level: " + isPalindrome("level"));       // true
        System.out.println("java: " + isPalindrome("java"));         // false
    }
}

Palindrome.main(null);
```
