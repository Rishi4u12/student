## Popcorn Hack: Trace Both Sorts

Hand predictions for `{31, 12, 47, 5, 26}`, written before running:

**Selection**
- Pass 1: smallest is 5, swap with index 0 → `[5, 12, 47, 31, 26]`
- Pass 2: 12 already in place → `[5, 12, 47, 31, 26]`
- Pass 3: smallest of the rest is 26, swap → `[5, 12, 26, 31, 47]`
- Pass 4: 31 already in place → `[5, 12, 26, 31, 47]`

**Insertion**
- Insert 12: shift 31 → `[12, 31, 47, 5, 26]`
- Insert 47: no shifts → `[12, 31, 47, 5, 26]`
- Insert 5: shift 47, 31, 12 → `[5, 12, 31, 47, 26]`
- Insert 26: shift 47, 31, stop at 12 → `[5, 12, 26, 31, 47]`

```java
import java.util.Arrays;

public class TraceBoth {
    public static void main(String[] args) {
        int[] original = {31, 12, 47, 5, 26};

        int[] sel = original.clone();
        System.out.println("Selection start: " + Arrays.toString(sel));
        for (int i = 0; i < sel.length - 1; i++) {
            int minIndex = i;
            for (int j = i + 1; j < sel.length; j++) {
                if (sel[j] < sel[minIndex]) {
                    minIndex = j;
                }
            }
            int temp = sel[i];
            sel[i] = sel[minIndex];
            sel[minIndex] = temp;
            System.out.println("Pass " + (i + 1) + ": " + Arrays.toString(sel));
        }

        int[] ins = original.clone();
        System.out.println("Insertion start: " + Arrays.toString(ins));
        for (int i = 1; i < ins.length; i++) {
            int key = ins[i];
            int j = i - 1;
            while (j >= 0 && ins[j] > key) {
                ins[j + 1] = ins[j];
                j--;
            }
            ins[j + 1] = key;
            System.out.println("Insert " + key + ": " + Arrays.toString(ins));
        }
    }
}

TraceBoth.main(null);
```

## Homework Hack: Sort the Garage

The assignment counts swaps and shifts. I also count comparisons, because that's where the two sorts differ on sorted input.

```java
import java.util.Arrays;

public class SortTheGarage {
    static int comparisons = 0;

    public static int selectionSort(int[] arr) {
        int swaps = 0;
        for (int i = 0; i < arr.length - 1; i++) {
            int minIndex = i;
            for (int j = i + 1; j < arr.length; j++) {
                comparisons++;
                if (arr[j] < arr[minIndex]) {
                    minIndex = j;
                }
            }
            if (minIndex != i) {
                int temp = arr[i];
                arr[i] = arr[minIndex];
                arr[minIndex] = temp;
                swaps++;
            }
            System.out.println("Pass " + (i + 1) + ": " + Arrays.toString(arr));
        }
        return swaps;
    }

    public static int insertionSort(int[] arr) {
        int shifts = 0;
        for (int i = 1; i < arr.length; i++) {
            int key = arr[i];
            int j = i - 1;
            while (j >= 0) {
                comparisons++;
                if (arr[j] <= key) {
                    break;
                }
                arr[j + 1] = arr[j];
                shifts++;
                j--;
            }
            arr[j + 1] = key;
            System.out.println("Insert " + key + ": " + Arrays.toString(arr));
        }
        return shifts;
    }

    public static void main(String[] args) {
        final String GARAGE_NAME = "Harbor Street Garage";
        int[] tickets = {305, 118, 442, 207, 96};

        int[] selCopy = tickets.clone();
        comparisons = 0;
        int swaps = selectionSort(selCopy);
        System.out.println(GARAGE_NAME + " selection: " + Arrays.toString(selCopy) + ", swaps: " + swaps + ", comparisons: " + comparisons);

        int[] insCopy = tickets.clone();
        comparisons = 0;
        int shifts = insertionSort(insCopy);
        System.out.println(GARAGE_NAME + " insertion: " + Arrays.toString(insCopy) + ", shifts: " + shifts + ", comparisons: " + comparisons);

        int[] sorted = {96, 118, 207, 305, 442};
        comparisons = 0;
        swaps = selectionSort(sorted.clone());
        System.out.println("Already sorted, selection: swaps " + swaps + ", comparisons " + comparisons);
        comparisons = 0;
        shifts = insertionSort(sorted.clone());
        System.out.println("Already sorted, insertion: shifts " + shifts + ", comparisons " + comparisons);
    }
}

SortTheGarage.main(null);
```

## Which Did Less Work on Sorted Input?

**Insertion Sort did less work.** Neither sort moved anything: 0 swaps and 0 shifts, so the fewer-moves difference is 0. The difference is in the comparisons, 4 for Insertion Sort and 10 for Selection Sort. Insertion Sort stops as soon as the value before the key isn't bigger, so it makes one comparison per insertion. Selection Sort always scans the whole unsorted part, because it can't know it has the minimum until it has looked at everything.
