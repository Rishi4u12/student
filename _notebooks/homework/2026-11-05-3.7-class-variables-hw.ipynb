## Popcorn Hack 1: Shared Counter

There's one `numberOfStudents` for the whole class, so each constructor call adds to the same variable.

```java
public class Student1 {
    private String name;
    public static int numberOfStudents = 0;

    public Student1(String name) {
        this.name = name;
        numberOfStudents++;
    }

    public static void main(String[] args) {
        Student1 a = new Student1("Alice");
        Student1 b = new Student1("Bob");
        Student1 c = new Student1("Charlie");
        System.out.println(Student1.numberOfStudents); // 3
    }
}

Student1.main(null);
```

## Popcorn Hack 2: Static `printName`

A static method has no current object, so `name` alone has no meaning. Taking a `Student2` parameter tells it whose name to read while keeping it `static`. The lesson's last line calls `Student.main(null)`, which is a typo for `Student2`.

```java
public class Student2 {
    private String name;

    public Student2(String name) {
        this.name = name;
    }

    public static void printName(Student2 student) {
        System.out.println(student.name);
    }

    public static void main(String[] args) {
        Student2 s = new Student2("Alice");
        Student2.printName(s);
    }
}

Student2.main(null);
```

## Homework: `BankAccount`

`resetBalance` subtracts this account's balance from the total before clearing it, so `totalBalance` stays correct. The cast in `getPercentOfMax` avoids integer division.

**Test results, from running the project's real JUnit tests on this code (with the `package` line restored):** 7 of 10 pass together. I don't believe the other three failures are bugs in this solution:

| Test | Result | Why |
|---|---|---|
| `testPercentOfMax` | Fails on any solution | `assertEquals(25.0, res)` resolves to JUnit's deprecated double overload, which always fails. It needs a delta, such as `assertEquals(25.0, res, 0.001)`. |
| `testAccountCount` | Passes alone, fails in the full run | The static counter keeps counting accounts from the other tests. |
| `testTotalBalance` | Passes alone, fails in the full run | The static total keeps deposits from the other tests. |

```java
public class BankAccount {
    public static final int MAX_BALANCE = 1000;

    private static int userCount = 0;
    private static int totalBalance = 0;

    private String username;
    private int balance;

    public BankAccount(String name) {
        username = name;
        balance = 0;
        userCount++;
    }

    public void deposit(int amount) {
        balance += amount;
        totalBalance += amount;
    }

    public int getBalance() {
        return balance;
    }

    public String getUsername() {
        return username;
    }

    public void resetBalance() {
        totalBalance -= balance;
        balance = 0;
    }

    public static int getUserCount() {
        return userCount;
    }

    public static int getTotalBalance() {
        return totalBalance;
    }

    public static void printAccount(BankAccount account) {
        System.out.println("Account: " + account.username);
        System.out.println("Balance: " + account.balance);
    }

    public static Double getPercentOfMax(BankAccount account) {
        return (double) account.balance / MAX_BALANCE * 100;
    }

    public static boolean hasReachedMax(BankAccount account) {
        return account.balance >= MAX_BALANCE;
    }

    public static void main(String[] args) {
        BankAccount alice = new BankAccount("Alice");
        BankAccount bob = new BankAccount("Bob");
        alice.deposit(250);
        bob.deposit(1000);
        BankAccount.printAccount(alice);
        System.out.println("Accounts: " + BankAccount.getUserCount());
        System.out.println("Total: " + BankAccount.getTotalBalance());
        System.out.println("Alice percent of max: " + BankAccount.getPercentOfMax(alice));
        System.out.println("Bob reached max: " + BankAccount.hasReachedMax(bob));
        bob.resetBalance();
        System.out.println("Total after Bob's reset: " + BankAccount.getTotalBalance());
    }
}

BankAccount.main(null);
```
