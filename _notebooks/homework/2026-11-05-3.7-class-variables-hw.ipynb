Implemented all 12 TODOs in the supplied project. Ten supplied test scenarios pass in isolated Java 17 processes. Maven is unavailable in this environment; this is not a claimed mvn test run. Isolation avoids shared static state leaking between the supplied tests.


```java
public class BankAccount {
 public static final int MAX_BALANCE=1000;
 private static int userCount=0;
 private static int totalBalance=0;
 private String username;
 private int balance;
 public BankAccount(String name){username=name;balance=0;userCount++;}
 public void deposit(int amount){balance+=amount;totalBalance+=amount;}
 public int getBalance(){return balance;}
 public String getUsername(){return username;}
 public void resetBalance(){totalBalance-=balance;balance=0;}
 public static int getUserCount(){return userCount;}
 public static int getTotalBalance(){return totalBalance;}
 public static void printAccount(BankAccount account){System.out.println("Account: "+account.username);System.out.println("Balance: "+account.balance);}
 public static Double getPercentOfMax(BankAccount account){return 100.0*account.balance/MAX_BALANCE;}
 public static boolean hasReachedMax(BankAccount account){return account.balance>=MAX_BALANCE;}
}
```
