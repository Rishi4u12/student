## Popcorn Hack: Countdown

The condition `start >= 1` ends the loop, and `start--` moves toward that end on every pass. Prediction: 5, 4, 3, 2, 1, then `Liftoff!`.

```java
public class Countdown {
    public static void main(String[] args) {
        int start = 5;

        while (start >= 1) {
            System.out.println(start);
            start--;
        }
        System.out.println("Liftoff!");
    }
}

Countdown.main(null);
```

## Homework Hack: Doubling

Trace: 100 → 200 → 400 → 800 → 1600, which is 4 doublings. The loop stops at 1600 because that's the first value not below 1000.

```java
public class Doubling {
    public static void main(String[] args) {
        int amount = 100;
        int goal = 1000;
        int doublings = 0;

        while (amount < goal) {
            amount *= 2;
            doublings++;
        }

        System.out.println("Amount: " + amount);          // expected: 1600
        System.out.println("Doublings: " + doublings);    // expected: 4
    }
}

Doubling.main(null);
```
