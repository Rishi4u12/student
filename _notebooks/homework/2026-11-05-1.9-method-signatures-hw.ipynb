## Popcorn Hack: Signature Detective

```java
public class SignatureDetective {
    String sig1 = "getScore(String)";
    String sig2 = "average(double, double)";
    String sig3 = "reset()";

    public static String pick(int x)        { return "A"; }
    public static String pick(double x)     { return "B"; }
    public static String pick(int x, int y) { return "C"; }
    public static String pick(String s)     { return "D"; }

    String guess1 = "A";  // pick(4): exact match on int
    String guess2 = "B";  // pick(4.0): exact match on double
    String guess3 = "C";  // pick(4, 5): the only two-parameter version
    String guess4 = "D";  // pick("4"): the argument is a String

    public static void main(String[] args) {
        SignatureDetective me = new SignatureDetective();
        System.out.println(me.sig1 + ", " + me.sig2 + ", " + me.sig3);
        System.out.println("Guesses: " + me.guess1 + me.guess2 + me.guess3 + me.guess4);
        System.out.println("Actual:  " + pick(4) + pick(4.0) + pick(4, 5) + pick("4"));
    }
}

SignatureDetective.main(null);
```

## Popcorn Hacks: Overloads

```java
public class PopcornOverloads {
    public static int max(int a, int b) {
        return a >= b ? a : b;
    }

    public static double max(double a, double b) {
        return a >= b ? a : b;
    }

    static void print(int n) {
        System.out.println("int:" + n);
    }

    static void print(String s) {
        System.out.println("str:" + s);
    }

    public static void main(String[] args) {
        System.out.println(max(3, 9));      // 9
        System.out.println(max(-2, -7));    // -2
        System.out.println(max(3.5, 2.9));  // 3.5
        System.out.println(max(2, 2.0));    // 2.0, the int widens so the double version runs
        print(42);
        print("hello");
        print('A' + "!");
    }
}

PopcornOverloads.main(null);
```

## Homework: Area Overloads

`area(3)` calls `area(double)`, because no `area(int)` exists and the `int` widens to `double`.

```java
public class HomeworkStarter {
    // Signature: area(double)
    public static double area(double radius) {
        return Math.PI * radius * radius;
    }

    // Signature: area(int, int)
    public static int area(int width, int height) {
        return width * height;
    }

    public static void main(String[] args) {
        System.out.println(area(3.0));
        System.out.println(area(3, 4));
        System.out.println(area(3));
    }
}

HomeworkStarter.main(null);
```

## Coding Tasks

Task 3 predictions: `show(7)` prints `int` (exact match), `show(7L)` prints `long` (exact match), and `show(7.0)` prints `double` (exact match).

```java
public class CodingTasks {
    static int abs(int x) { return x < 0 ? -x : x; }
    static double abs(double x) { return x < 0 ? -x : x; }
    static long abs(long x) { return x < 0 ? -x : x; }

    static String concat(String a, String b) { return a + b; }

    static String concat(String a, int n) {
        String result = "";
        for (int i = 0; i < n; i++) {
            result += a;
        }
        return result;
    }

    static void show(int x) { System.out.println("int"); }
    static void show(double x) { System.out.println("double"); }
    static void show(long x) { System.out.println("long"); }

    public static void main(String[] args) {
        System.out.println(abs(-4) + " " + abs(-4.5) + " " + abs(-4000000000L));
        System.out.println(concat("AP", "CSA") + " " + concat("ha", 3));
        show(7);
        show(7L);
        show(7.0);
    }
}

CodingTasks.main(null);
```

## FRQ-Style

Assumptions: `s` and `target` are not null, and an empty `String` target is found at index 0. `clamp` swaps `low` and `high` first if they're reversed.

```java
public class FrqOverloads {
    public static int indexOf(char target, String s) {
        for (int i = 0; i < s.length(); i++) {
            if (s.charAt(i) == target) {
                return i;
            }
        }
        return -1;
    }

    public static int indexOf(String target, String s) {
        for (int i = 0; i + target.length() <= s.length(); i++) {
            if (s.substring(i, i + target.length()).equals(target)) {
                return i;
            }
        }
        return -1;
    }

    public static int clamp(int value, int low, int high) {
        if (low > high) {
            int temp = low;
            low = high;
            high = temp;
        }
        if (value < low) return low;
        if (value > high) return high;
        return value;
    }

    public static double clamp(double value, double low, double high) {
        if (low > high) {
            double temp = low;
            low = high;
            high = temp;
        }
        if (value < low) return low;
        if (value > high) return high;
        return value;
    }

    public static void main(String[] args) {
        System.out.println(indexOf('v', "java"));       // 2
        System.out.println(indexOf('z', "java"));       // -1
        System.out.println(indexOf("va", "java"));      // 2
        System.out.println(indexOf("java", "ja"));      // -1
        System.out.println(clamp(15, 0, 10));           // 10
        System.out.println(clamp(5, 10, 0));            // 5, bounds swapped
        System.out.println(clamp(-1.5, 0.0, 1.0));      // 0.0
    }
}

FrqOverloads.main(null);
```
