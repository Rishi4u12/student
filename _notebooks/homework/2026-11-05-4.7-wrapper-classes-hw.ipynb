## Popcorn Hack: Receipt From Text

- **Primitive values from parsing:** `Integer.parseInt` gives an `int`, and `Double.parseDouble` gives a `double`.
- **Where boxing occurs:** at `return total;`, where the `double` is autoboxed to match the `Double` return type.
- **My additional input:** `"4,0.5"`, expected result `2.0`.

```java
public class ReceiptPractice {
    public static Double receiptTotal(String record) {
        String[] fields = record.split(",");
        int quantity = Integer.parseInt(fields[0]);
        double unitPrice = Double.parseDouble(fields[1]);
        double total = quantity * unitPrice;
        return total;  // autoboxed to Double
    }

    public static void main(String[] args) {
        String[] records = {"3,2.5", "0,9.99", "2,1.25", "4,0.5"};
        double[] expected = {7.5, 0.0, 2.5, 2.0};
        for (int i = 0; i < records.length; i++) {
            Double result = receiptTotal(records[i]);
            boolean pass = result != null && Math.abs(result - expected[i]) < 0.000001;
            System.out.println((pass ? "PASS" : "FAIL") + " " + records[i]
                + ": expected " + expected[i] + ", actual " + result);
        }
    }
}

ReceiptPractice.main(null);
```

## Exit Ticket

1. `double`.
2. The `String` is parsed into an `int`, and that `int` is autoboxed into an `Integer` for the assignment.
3. `score + 1` unboxes to an `int`, adds, and boxes the result into a new `Integer`. The variable is reassigned, and the original wrapper object's value never changes.
4. `82`, because `+` with a `String` concatenates. Then `10`, because `parseInt` produces a number first.

## Prediction

Two receipts, `3,2.5` and `2,1.25`, should total 7.5 + 2.5 = 10.0. An empty array should return 0.0 because the loop never runs. `3 * 0.1` may print as `0.30000000000000004`, since `double` can't represent 0.1 exactly, which is why the check uses a tolerance.

On wrapper immutability: `Integer score = 4; score = score + 1;` makes `score` refer to a new `Integer` object, and the first object still holds 4.

## Solution

- **Parsing:** `Integer.parseInt` and `Double.parseDouble` inside the loop.
- **Accumulating:** `total` is a primitive `double`.
- **Autoboxing:** `return total;` converts it to `Double`.
- **Unboxing:** `actual - expected` in `check` converts the `Double` back to a `double` so it can be subtracted.

```java
public class SnackShopHomework {
    public static Double orderTotal(String[] records) {
        double total = 0.0;
        for (String record : records) {
            String[] fields = record.split(",");
            int quantity = Integer.parseInt(fields[0]);
            double unitPrice = Double.parseDouble(fields[1]);
            total += quantity * unitPrice;
        }
        return total;  // autoboxed to Double
    }

    public static void check(String label, String[] records, double expected) {
        Double actual = orderTotal(records);
        boolean pass = actual != null && Math.abs(actual - expected) < 0.000001;
        System.out.println((pass ? "PASS" : "FAIL") + " " + label
            + ": expected " + expected + ", actual " + actual);
    }

    public static void main(String[] args) {
        check("one receipt", new String[] {"3,2.5"}, 7.5);
        check("two receipts", new String[] {"3,2.5", "2,1.25"}, 10.0);
        check("zero quantity", new String[] {"0,9.99"}, 0.0);
        check("empty order", new String[] {}, 0.0);
        check("decimal price", new String[] {"3,0.1"}, 0.3);
        check("my test: three records, one free", new String[] {"2,1.5", "1,0.0", "4,0.25"}, 4.0);
    }
}

SnackShopHomework.main(null);
```
