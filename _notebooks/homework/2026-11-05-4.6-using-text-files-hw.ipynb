## Popcorn Hack: Count Busy Days

**My trace for the boundary value 20:** `hasNext()` is true, `nextInt()` reads 20, and `20 >= 20` is true, so `busyDays` becomes 1. `hasNext()` is then false and the method returns 1.

**One test I would add:** `"20 20 19"` with expected result 2, to confirm repeated boundary values each count.

```java
import java.io.File;
import java.io.IOException;
import java.io.PrintWriter;
import java.util.Scanner;

public class BusyDaysPractice {
    public static File sampleFile(String text) throws IOException {
        File file = File.createTempFile("csa-attendance-", ".txt");
        file.deleteOnExit();
        PrintWriter writer = new PrintWriter(file);
        writer.print(text);
        writer.close();
        return file;
    }

    public static int countBusyDays(String filename) throws IOException {
        Scanner input = new Scanner(new File(filename));
        int busyDays = 0;
        while (input.hasNext()) {
            int attendance = input.nextInt();
            if (attendance >= 20) {
                busyDays++;
            }
        }
        input.close();
        return busyDays;
    }

    public static void main(String[] args) throws IOException {
        String[] data = {"18 24 21 17", "20", "19 0", "", "20 20 19"};
        int[] expected = {2, 1, 0, 0, 2};
        for (int i = 0; i < data.length; i++) {
            int actual = countBusyDays(sampleFile(data[i]).getPath());
            System.out.println((actual == expected[i] ? "PASS" : "FAIL")
                + " case " + i + ": expected " + expected[i] + ", actual " + actual);
        }
    }
}

BusyDaysPractice.main(null);
```

## Exit Ticket

1. `hasNext()` checks whether another token exists without moving forward. `nextInt()` consumes the next token as an `int`, and it can fail if the token isn't a valid integer or nothing is left.
2. It compiles because the compiler doesn't check the file system. It can fail at run time if the path doesn't lead to a readable file from the working directory.
3. The `String` `"Blue"`.

## Prediction

Tracing `18 24 21 17` with threshold 20: 18 is skipped, 24 makes the total 24, 21 makes it 45, and 17 is skipped. The result is 45. An answer of 2 would be the *count* of busy days, which answers a different question from total attendance on those days.

An empty file returns 0 because `hasNext()` is false right away, so the loop body never runs and `total` keeps its starting value.

## Solution

The line that advances the scanner is `int attendance = input.nextInt();`. Without it, `hasNext()` would stay true forever.

```java
import java.io.File;
import java.io.IOException;
import java.io.PrintWriter;
import java.util.Scanner;

public class AttendanceHomework {
    public static File sampleFile(String text) throws IOException {
        File file = File.createTempFile("csa-attendance-", ".txt");
        file.deleteOnExit();
        PrintWriter writer = new PrintWriter(file);
        writer.print(text);
        writer.close();
        return file;
    }

    public static int totalAtLeast(String filename, int threshold) throws IOException {
        Scanner input = new Scanner(new File(filename));
        int total = 0;
        while (input.hasNext()) {
            int attendance = input.nextInt();
            if (attendance >= threshold) {
                total += attendance;
            }
        }
        input.close();
        return total;
    }

    public static void main(String[] args) throws IOException {
        // The last case is my own test: a different threshold, across several lines.
        String[] data = {"18 24 21 17", "20", "19 0", "", "20\n20 21", "5\n0\n30"};
        int[] thresholds = {20, 20, 20, 20, 20, 5};
        int[] expected = {45, 20, 0, 0, 61, 35};
        for (int i = 0; i < data.length; i++) {
            int actual = totalAtLeast(sampleFile(data[i]).getPath(), thresholds[i]);
            System.out.println((actual == expected[i] ? "PASS" : "FAIL")
                + " case " + i + ": expected " + expected[i] + ", actual " + actual);
        }
    }
}

AttendanceHomework.main(null);
```

## Tests
