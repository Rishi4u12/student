## Vocabulary Filter Sandbox

The backward loop is safe because removing an element only shifts elements that have already been checked. The list includes adjacent short words, which is the case a forward loop would miss.

```java
import java.util.ArrayList;
import java.util.Arrays;
import java.util.ConcurrentModificationException;

public class WordFilter {
    /**
     * Removes all words with fewer than 4 characters from the list.
     * Uses backward loop traversal.
     */
    public static void removeShortWords(ArrayList<String> words) {
        for (int i = words.size() - 1; i >= 0; i--) {
            if (words.get(i).length() < 4) {
                words.remove(i);
            }
        }
    }

    public static void main(String[] args) {
        ArrayList<String> words = new ArrayList<String>(Arrays.asList("cat", "dog", "elephant", "of", "java", "loop"));
        removeShortWords(words);
        System.out.println(words);   // expect [elephant, java, loop]

        ArrayList<String> others = new ArrayList<String>(Arrays.asList("a", "bbbb", "c", "dddd"));
        try {
            for (String w : others) {
                if (w.length() < 4) {
                    others.remove(w);
                }
            }
        } catch (ConcurrentModificationException error) {
            System.out.println("Enhanced for threw: " + error.getClass().getSimpleName());
        }
    }
}

WordFilter.main(null);
```

## Short Answer Reflection

An enhanced `for` loop over an `ArrayList` uses an iterator behind the scenes. Calling `words.remove(w)` changes the list's size directly, without going through that iterator. On its next step, the iterator notices the list was structurally modified since it started and throws `ConcurrentModificationException` instead of risking wrong results. The second part of the cell above shows this happening.
