`songs.size() - 1` finds the last index even after inserting at index 0 shifted everything right.

```java
import java.util.ArrayList;

public class MyPlaylist {
    public static void main(String[] args) {
        ArrayList<String> songs = new ArrayList<String>();
        songs.add("Song A");
        songs.add("Song B");
        songs.add("Song C");
        songs.add(0, "Favorite Song");
        songs.set(songs.size() - 1, "Song D");
        songs.remove("Song A");
        System.out.println(songs + ", " + songs.size() + " songs");
    }
}

MyPlaylist.main(null);
```

## Homework Hack: List Tools

`temp` saves the first value before `set` overwrites it. `Integer.valueOf(target)` matters because `nums.remove(7)` with a plain `int` would mean "remove index 7," while an `Integer` selects `remove(Object obj)`.

```java
import java.util.ArrayList;

public class ListTools {
    public static void swapFirstAndLast(ArrayList<Integer> nums) {
        int temp = nums.get(0);
        nums.set(0, nums.get(nums.size() - 1));
        nums.set(nums.size() - 1, temp);
    }

    public static void removeFirstOccurrence(ArrayList<Integer> nums, int target) {
        nums.remove(Integer.valueOf(target));
    }

    public static void main(String[] args) {
        ArrayList<Integer> nums = new ArrayList<Integer>();
        nums.add(1); nums.add(7); nums.add(3); nums.add(7);
        swapFirstAndLast(nums);
        System.out.println(nums);   // expect [7, 7, 3, 1]
        removeFirstOccurrence(nums, 7);
        System.out.println(nums);   // expect [7, 3, 1]
    }
}

ListTools.main(null);
```
