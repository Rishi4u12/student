## Popcorn Hack: Pet

The fields are `private`, so outside code has to use the public methods. `stage()` is a `private` helper because only `describe()` needs it.

```java
public class Pet {
    private String name;
    private int age;

    public Pet(String name, int age) {
        this.name = name;
        this.age = age;
    }

    public String getName() {
        return name;
    }

    public void birthday() {
        age++;
    }

    private String stage() {
        return age < 3 ? "young" : "grown";
    }

    public String describe() {
        return name + " is " + age + " and " + stage();
    }

    public static void main(String[] args) {
        Pet pet = new Pet("Pepper", 2);
        System.out.println(pet.describe());
        pet.birthday();
        System.out.println(pet.getName());
        System.out.println(pet.describe());
    }
}

Pet.main(null);
```

## Homework Hack: Book

`Math.min` keeps `pagesRead` from going past `totalPages`, which is why `readPages(500)` stops at the book's length. `isFinished` is private because outside code only needs `getStatus`.

```java
public class Book {
    private String title;
    private int totalPages;
    private int pagesRead;

    public Book(String title, int totalPages) {
        this.title = title;
        this.totalPages = totalPages;
        this.pagesRead = 0;
    }

    public int getPagesRead() {
        return pagesRead;
    }

    public void readPages(int pages) {
        pagesRead = Math.min(pagesRead + pages, totalPages);
    }

    private boolean isFinished() {
        return pagesRead == totalPages;
    }

    public String getStatus() {
        return isFinished() ? "Finished" : "Reading";
    }

    public static void main(String[] args) {
        Book book = new Book("Dune", 412);
        book.readPages(100);
        System.out.println(book.getStatus());      // Reading
        book.readPages(500);
        System.out.println(book.getPagesRead());   // 412
        System.out.println(book.getStatus());      // Finished
    }
}

Book.main(null);
```
