## Homework Hack: Shipping

The `if-else` picks exactly one shipping line. The total prints after the branch, so it runs either way. Using `>=` puts an order of exactly $50 in the free shipping branch.

Prediction: 42.50 pays shipping, while 50.00 and 75.00 ship free.

```java
public class Shipping {
    public static void printOrder(double orderTotal) {
        if (orderTotal >= 50) {
            System.out.println("Free shipping");
        } else {
            System.out.println("Shipping: $5.99");
        }
        System.out.println("Order total: $" + orderTotal);
    }

    public static void main(String[] args) {
        printOrder(42.50);
        printOrder(50.00);
        printOrder(75.00);
    }
}

Shipping.main(null);
```
