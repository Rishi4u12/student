## Popcorn Hack: Two Bugs

Bug 1: `=>` isn't an operator, so `>=` is what compiles. Bug 2: `!=` asks "not equal," but the comment asks for "exactly equal," so it should be `==`. Prediction: `true`, then `false`, since 120 is at least 100 and isn't equal to it.

```java
public class BonusCheck {
    public static void main(String[] args) {
        int score = 120;
        int target = 100;

        boolean bonus = score >= target;   // Bug 1 fixed: was =>
        boolean exact = score == target;   // Bug 2 fixed: was !=

        System.out.println("Bonus unlocked: " + bonus);
        System.out.println("Exact score: " + exact);
    }
}

BonusCheck.main(null);
```

## Homework Hack: Speed Check

A relational operator produces a `boolean` directly, so each result can be stored without an `if`. `speed % 2 != 0` means dividing by 2 leaves a remainder, which makes the number odd.

```java
public class SpeedCheck {
    public static void main(String[] args) {
        int speed = 68;
        int limit = 65;
        int age = 17;

        boolean speeding = speed > limit;
        boolean exactLimit = speed == limit;
        boolean adult = age >= 18;
        boolean oddSpeed = speed % 2 != 0;

        System.out.println("speeding: " + speeding);      // expected: true
        System.out.println("exactLimit: " + exactLimit);  // expected: false
        System.out.println("adult: " + adult);            // expected: false
        System.out.println("oddSpeed: " + oddSpeed);      // expected: false
    }
}

SpeedCheck.main(null);
```
