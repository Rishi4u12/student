## Concepts

An array of length `n` has valid indices `0` through `n - 1`, because counting starts at 0. A standard `for` loop fits when a position matters, such as a rank. An enhanced `for` loop fits when only the values matter. It doesn't modify the array, because its loop variable holds a copy of each element.

## Popcorn Hack: Leaderboard

```java
public class LeaderboardPopcorn {
    public static void main(String[] args) {
        int[] scores = {640, 982, 755, 871, 590, 913};

        System.out.println("Ranked players:");
        for (int rank = 0; rank < scores.length; rank++) {
            System.out.println("Rank " + (rank + 1) + ": " + scores[rank]);
        }

        System.out.println("Score values:");
        for (int score : scores) {
            System.out.println(score);
        }
    }
}

LeaderboardPopcorn.main(null);
```

## Homework Hack: Off-by-One

First, the starter with its bug. `<=` lets `rank` reach 5, which isn't a valid index, so the catch block runs after the fifth line.

```java
public class LeaderboardBuggy {
    public static void main(String[] args) {
        int[] scores = {710, 865, 903, 947, 981};
        try {
            for (int rank = 0; rank <= scores.length; rank++) {
                System.out.println("Rank " + (rank + 1) + ": " + scores[rank]);
            }
        } catch (ArrayIndexOutOfBoundsException error) {
            System.out.println("Repair needed: " + error.getMessage());
        }
    }
}

LeaderboardBuggy.main(null);
```

An array of length 5 has valid indices 0 through 4, so `<` is right: it stops the loop when `rank` equals the length. I kept a standard `for` loop because the rank is printed.

**Second task:** the array now has seven scores and the condition is unchanged. `scores.length` updates automatically, so every rank still prints.
