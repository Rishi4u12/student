## Popcorn Hack: Number Triangle

The inner loop's bound is `i`, so each row is one number longer. `print` keeps a row together and `println` ends it.

```java
public class NumberTriangle {
    public static void main(String[] args) {
        int rows = 4;

        for (int i = 1; i <= rows; i++) {
            for (int j = 1; j <= i; j++) {
                System.out.print(j + " ");
            }
            System.out.println();
        }
    }
}

NumberTriangle.main(null);
```

## Homework Hack: Even-Sum Pairs

Starting `b` at `a + 1` makes `a < b` always true, so no pair is counted twice. Prediction: `n = 4` gives 2, `n = 5` gives 4, and `n = 6` gives 6.

```java
public class EvenPairs {
    public static void countPairs(int n) {
        int pairs = 0;
        for (int a = 1; a <= n; a++) {
            for (int b = a + 1; b <= n; b++) {
                if ((a + b) % 2 == 0) {
                    pairs++;
                }
            }
        }
        System.out.println("n = " + n + ", even-sum pairs: " + pairs);
    }

    public static void main(String[] args) {
        countPairs(4);
        countPairs(5);
        countPairs(6);
    }
}

EvenPairs.main(null);
```
