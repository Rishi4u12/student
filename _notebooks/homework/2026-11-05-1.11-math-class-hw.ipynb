## Popcorn Hack 1: Math Predictions

Predictions: `Math.abs(-8)` is `8`, `Math.pow(3, 2)` is `9.0`, and `Math.sqrt(36)` is `6.0`. `pow` and `sqrt` return `double`, so they print with a decimal.

```java
public class PopcornMath {
    static String guess1 = "8";
    static String guess2 = "9.0";
    static String guess3 = "6.0";

    public static void main(String[] args) {
        System.out.println("Predictions: " + guess1 + ", " + guess2 + ", " + guess3);
        System.out.println("Actual:");
        System.out.println(Math.abs(-8));
        System.out.println(Math.pow(3, 2));
        System.out.println(Math.sqrt(36));

        double distance = Math.sqrt(Math.pow(3 - 0, 2) + Math.pow(4 - 0, 2));
        System.out.println("Distance: " + distance);
    }
}

PopcornMath.main(null);
```

## Popcorn Hack 2: Random Range

Instead of rerunning by hand, this rolls 1,000 times and reports the lowest and highest values seen, which should never be 9 or 21.

```java
public class PopcornRange {
    public static void main(String[] args) {
        int min = 10;
        int max = 20;
        int lowest = max;
        int highest = min;
        for (int i = 0; i < 1000; i++) {
            int roll = (int) (Math.random() * (max - min + 1)) + min;
            lowest = Math.min(lowest, roll);
            highest = Math.max(highest, roll);
        }
        System.out.println("Lowest roll: " + lowest + ", highest roll: " + highest);
    }
}

PopcornRange.main(null);
```

## Homework Hack: Loot Drop

```java
public class RandomLoot {
    public static void main(String[] args) {
        int rarityRoll = (int) (Math.random() * 100) + 1;

        String rarity;
        int gold;
        if (rarityRoll <= 60) {
            rarity = "Common";
            gold = (int) (Math.random() * 21) + 10;   // 10-30
        } else if (rarityRoll <= 85) {
            rarity = "Rare";
            gold = (int) (Math.random() * 40) + 31;   // 31-70
        } else {
            rarity = "Legendary";
            gold = (int) (Math.random() * 30) + 71;   // 71-100
        }

        System.out.println("Loot Drop!");
        System.out.println("Rarity Roll: " + rarityRoll);
        System.out.println("You got: " + rarity + " item");
        System.out.println("Gold Value: " + gold);
    }
}

RandomLoot.main(null);
```

## Homework Page: Game Stats Calculator

```java
public class GameStats {
    public static int healthDifference(int player1Health, int player2Health) {
        return Math.abs(player1Health - player2Health);
    }

    public static double calculateDamage(double baseDamage, double powerLevel) {
        return baseDamage * Math.pow(1.5, powerLevel);
    }

    public static double findDistance(int playerX, int playerY, int enemyX, int enemyY) {
        return Math.sqrt(Math.pow(enemyX - playerX, 2) + Math.pow(enemyY - playerY, 2));
    }

    public static int generateLoot(int minValue, int maxValue) {
        return (int) (Math.random() * (maxValue - minValue + 1)) + minValue;
    }

    public static void main(String[] args) {
        System.out.println(healthDifference(75, 120));      // 45
        System.out.println(healthDifference(100, 80));      // 20
        System.out.println(healthDifference(50, 50));       // 0
        System.out.println(calculateDamage(10.0, 2));       // 22.5
        System.out.println(calculateDamage(15.0, 3));       // 50.625
        System.out.println(findDistance(0, 0, 3, 4));       // 5.0
        System.out.println(findDistance(1, 1, 4, 5));       // 5.0
        System.out.println(generateLoot(100, 100));         // 100
        System.out.println("Dice roll: " + generateLoot(1, 6));
    }
}

GameStats.main(null);
```
