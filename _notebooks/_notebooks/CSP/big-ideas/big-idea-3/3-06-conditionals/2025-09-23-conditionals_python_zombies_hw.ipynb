```python
# Task 1: Simple if/elif/else
# Input: player's health and nearby zombies; Output: action string
health = 35
zombies_nearby = 3

if health <= 0:
    action = "game over"
elif health < 20:
    action = "hide"
elif zombies_nearby >= 5:
    action = "run"
else:
    action = "fight"

print(action)  # expected: fight
```

```python
# Task 2: Nested conditionals
# Rule: If inventory has 'medkit', use it when health < 30; otherwise avoid fights with >=3 zombies
health = 25
inventory = ["key", "medkit", "map"]
zombies_nearby = 4

if health < 30:
    if "medkit" in inventory:
        health += 30
        action = "heal"
    elif zombies_nearby >= 3:
        action = "avoid"
    else:
        action = "hide"
else:
    action = "explore"

print(action, health)  # expected: heal 55
```
