## 💎 Diamond Hacks(optional)

Create a function that takes an object with decision options and their desired probabilities (as percentages), and returns a randomly selected decision based on those weights.


````python
import random

def weighted_decision(options):
    """
    Takes a dictionary of options and their probabilities (as percentages),
    and returns a randomly selected option based on those weights.
    Example: {'Yes': 50, 'No': 30, 'Maybe': 20}
    """
    choices = list(options.keys())
    weights = list(options.values())
    return random.choices(choices, weights=weights, k=1)[0]

# Example usage:
decisions = {'Yes': 50, 'No': 30, 'Maybe': 20}
print("Weighted decision:", weighted_decision(decisions))
````

````python
import random
import string

# Creative Hack Example:
# Create a random password generator that generates a password of length n
# using uppercase, lowercase, digits, and special characters.


def generate_password(n):
    chars = string.ascii_letters + string.digits + "!@#$%^&*()"
    return ''.join(random.choice(chars) for _ in range(n))

# Example usage:
print("Random password:", generate_password(12))
````
