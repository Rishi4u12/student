```python
from random import random
import ipywidgets as widgets
from IPython.display import display, clear_output

# Task 1: Simple Loot Box
def open_loot_boxes(chance):
    rare_count = 0
    common_count = 0
    results = []
    for i in range(10):
        if random() < chance / 100:
            rare_count += 1
            results.append(f"Loot Box {i+1}: Rare Item")
        else:
            common_count += 1
            results.append(f"Loot Box {i+1}: Common Item")
    return rare_count, common_count, results

def show_loot_box_results(change):
    clear_output(wait=True)
    display(slider)
    rare, common, result_list = open_loot_boxes(slider.value)
    print(f"Chance of Rare: {slider.value}%")
    print(f"Rare Items: {rare}")
    print(f"Common Items: {common}")
    print("Results:")
    for r in result_list:
        print(r)

slider = widgets.IntSlider(value=30, min=10, max=90, step=1, description='Rare Chance (%)')
slider.observe(show_loot_box_results, names='value')
display(slider)
show_loot_box_results(None)
```
