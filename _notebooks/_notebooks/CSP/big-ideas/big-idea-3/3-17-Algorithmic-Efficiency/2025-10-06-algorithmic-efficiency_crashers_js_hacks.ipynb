### Hack 1: How Much Time?

#### Objective: write the time complexity of the algorithm below using Big-O notation.
(don't worry about special cases such as n = 1 or n = 0).

```python
%%javascript
let n = 10; // change this value to test different outputs!

for (let i = 0; i < n * 2; i++) {
    console.log(i);
}

// Print the algorithm's time complexity.
// Coefficients don't matter for Big-O, so n * 2 is still O(n).
console.log('Time complexity: O(n)');
```

### Hack 2: Your Turn!

#### Objective: <strong>write</strong> an algorithm with O(n^2) time complexity.

```python
%%javascript
const n = 10; // change this if you want.

// O(n^2) algorithm: nested loops that do constant work per pair (i, j)
let count = 0;
for (let i = 0; i < n; i++) {
	for (let j = 0; j < n; j++) {
		// constant-time operation
		console.log(`i=${i}, j=${j}`);
		count++;
	}
}
console.log('Total pairs:', count);
```

### Hack 4: Extra Challenge 

#### Objective: Write an algorithm that does <strong>NOT</strong> have a time complexity of O(1), O(n), or O(n^x) and identify the time complexity
##### (I will not accept O(n^3) or some other power, it needs to be more complex.)

```python
n = int(input())

# Validate input
if n < 0:
	raise ValueError("n must be non-negative")

# For safety in a notebook, cap n so we don't try to enumerate enormous numbers of subsets.
MAX_N = 20
if n > MAX_N:
	print(f"n is {n}, too large to enumerate 2^n subsets safely. Using n={MAX_N} instead for demonstration.")
	n = MAX_N

# Example algorithm with time complexity O(2^n): enumerate all subsets (the power set)
subsets = []
for mask in range(1 << n):  # 0 .. 2^n - 1
	subset = []
	for i in range(n):
		if (mask >> i) & 1:
			subset.append(i)
	subsets.append(subset)

print("Number of subsets:", len(subsets))
print("Time complexity: O(2^n)")
```
