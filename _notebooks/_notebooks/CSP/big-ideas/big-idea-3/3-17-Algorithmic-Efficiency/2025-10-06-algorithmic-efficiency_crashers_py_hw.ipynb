### Hack 1: How Much Time?

#### Objective: write the time complexity of the algorithm below using Big-O notation.
(don't worry about special cases such as n = 1 or n = 0).

```python
n = int(input()) # remember what O(n) means? This is a good way of visualizing n.

for i in range(n):
    print(i)

# print the above algorithm's time complexity
print("Time complexity: O(n)")
```

### Hack 2: Your Turn!

#### Objective: <strong>write</strong> an algorithm with O(n^2) time complexity.

```python
n = int(input())

# O(n^2) algorithm using nested loops
count = 0
for i in range(n):
	for j in range(n):
		# constant-time work per iteration
		count += 1

print(count)  # should print n*n
```

### Hack 3: Gotta Go Fast!

#### Objective: Optimize this algorithm so that it has a lower time complexity <strong>without modifying the outer loop</strong>

```python
import math
n = int(input())
count = 0

# compute how many times the inner loop would run for each i
inner_count = math.ceil(math.sqrt(n) * 2)

for i in range(n):
    # add the inner_count directly instead of iterating
    count += inner_count

print(count)

#TODO: make this algorithm more efficient, but keep the outer loop and make sure the output is still the same!
#Hint: how does the inner loop affect time complexity?
```

### Hack 4: Extra Challenge 

#### Objective: Write an algorithm that does <strong>NOT</strong> have a time complexity of O(1), O(n), or O(n^x) and identify the time complexity
##### (I will not accept O(n^3) or some other power, it needs to be more complex.)

```python
n = int(input())

# Exponential-time algorithm example: generate all subsets of an n-element set.
# This runs in O(n * 2^n) time (more complex than any polynomial O(n^x)).
# For safety, limit n so the notebook doesn't hang for very large inputs.
MAX_N = 20
if n < 0:
	raise ValueError("n must be non-negative")
if n > MAX_N:
	print(f"Input n={n} is large for an exponential demo; limiting to n={MAX_N}")
	n = MAX_N

count = 0
total_elements_across_subsets = 0

# Iterate over all 2^n subset masks
for mask in range(1 << n):
	# build the subset (takes O(n) per mask)
	subset = []
	for i in range(n):
		if mask & (1 << i):
			subset.append(i)
	count += 1
	total_elements_across_subsets += len(subset)

print("Number of subsets generated:", count)
print("Total number of elements across all subsets:", total_elements_across_subsets)
print("Time complexity: O(n * 2^n) — exponential")
```
