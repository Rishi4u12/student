## Q1 (Exponents):
### A cube has a side length of 6 units. What is its volume?

```python
// Q1 (Exponents): Calculate the volume of a cube with side length 6 units
let sideLength = 6;
let volume = Math.pow(sideLength, 3);
console.log("Volume of the cube:", volume);
```

## Q3 (Algorithm): 

Write JavaScript  code where you define variables and run commands that find the values of operations you apply onto them

```python
// Q3 (Algorithm): Example of mathematical operations in JavaScript

let a = 10;
let b = 3;

// Addition
let sum = a + b;

// Subtraction
let difference = a - b;

// Multiplication
let product = a * b;

// Division
let quotient = a / b;

// Modulus (remainder)
let remainder = a % b;

// Exponentiation
let power = a ** b;

console.log("Sum:", sum);
console.log("Difference:", difference);
console.log("Product:", product);
console.log("Quotient:", quotient);
console.log("Remainder:", remainder);
console.log("Power:", power);
```
