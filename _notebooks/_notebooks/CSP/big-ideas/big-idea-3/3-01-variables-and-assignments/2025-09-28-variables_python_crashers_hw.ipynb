## Hack 3: Python - Multiple Assignment and Math Operators

```python
# Q3: Define variables and apply multiple operations
a, b = 12, 5

results = {
    "sum": a + b,
    "difference": a - b,
    "product": a * b,
    "true_division": a / b,
    "floor_division": a // b,
    "modulo": a % b,
    "power_a_to_b": a ** b,
    "pemdas_example": (a + 8) / 2 + (b ** 2),
}

for k, v in results.items():
    print(f"{k}: {v}")
```
